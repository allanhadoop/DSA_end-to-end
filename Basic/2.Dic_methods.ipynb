```python
#------------Dictionary methods ------------------
d = {
    120 : 'cow',
    130 : 'cat',
    140 : 'dog',
    150 : 'lion',
    160 : 'tiger',
    170 : 'bird'
}

print(d.items())
d.update({"170" : 'parrot'})
print(d.pop(120))
print(d.items())
print(d.values())
s = d.copy()
print(s.items())
```
