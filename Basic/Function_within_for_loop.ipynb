```python
#-------How to call function in conjuction with for loop
#--Create the correct connections
true_edges = {
    ("Home", "School"),
    ("School", "Hospital")
}
#--Create the connections we want to check
candidate_edges = {
    ("Home", "School"),
    ("Home", "Hospital")              # <<---This is wrong that we should identify
}

def normalize_edge(edge):
    return tuple(sorted(edge))       # sorted() puts the names in alphabetical order, and tuple() converts the sorted list back into a tuple.
print(normalize_edge(("School", "Home")))

#-----Now let's apply normalization to every connection
true_edges = {normalize_edge(e) for e in true_edges}   #<<--- Call function for each edge we get within for loop
candidate_edges = {normalize_edge(e) for e in candidate_edges}

print("Unexpected roads:", candidate_edges - true_edges)
print("Missing roads:", true_edges - candidate_edges)
```
