```python
from difflib import SequenceMatcher

text1 = "Patient received chemotherapy"
text2 = "Patient received radiotherapy"

similarity = SequenceMatcher(None, text1, text2).ratio()
# SequenceMatcher	Compares two sequences, such as strings.
# .ratio()	Returns a similarity score from 0.0 to 1.0.

print(similarity)
```
