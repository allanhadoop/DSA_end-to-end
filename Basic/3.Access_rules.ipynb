```python
#--- List of lists -  [[], [], []]
from collections.abc import ItemsView
data = [
    [101, "Sam"],                     # << -- [[], [], []]
    [102, "Tim"],
    [103, "Kim"]
]
for i in data:
    print(i)                          # <<----------------

print("----------Get indiviual values---")
for j in data:
    print(j[0], j[1])                 # <<----------------

print("----------nested for loop---")
for row in data:
    for value in row:                 # <<----------------
        print(value)
print("----------Unpacking---")
for id, name in data:
    print(id, name)
```

```python
#----------List of tuples → [(), (), ()]

data = [
    ("Rahul", "Pune"),            # <--- List of tuples → [(), (), ()]
    ("Priya", "Mumbai"),
    ("John", "Delhi")
]
print("----------Get indiviual values---")
for i in data:
    print(i[0], i[1])
print("----------Unpacking--------------")
for name, city in data:          # << -- Unpacking of tuple ia allowed 
    print(name, city)    
```

```python
#-------------List of dictionaries--[{}, {}, {}]----------
datadic = [
    {"sam": "diabetic"},
    {"tim": "heart failure"},
    {"kim": "diabetic"}
]

for i in datadic:                       #<< -- gets each dictionary.
    for name, disease in i.items():    #<<-- we use .items since we need key and value 
        print(name, disease)
```

```python
#------List of dictionaries containing lists
Key1 = [
    {"disease": "diabetic", 
     "Key2": ["Sam", "Kim"]},            # << ---- List within dictionary
   
    {"disease": "heart failure", 
     "Key2": ["Tim", "Bim"]}
]

for i in Key1:                
    print(i["disease"])

    for name in i["Key2"]:
        print(name)
```

```python
#----------Dictionary containing lists-------
data = {                                       # <<--- This is dictionary
    "diabetic": ["Sam", "Kim"],
    "heart failure": ["Tim", "Bim"]
}
for disease, patients in data.items():         # << -- Since this is dic, we will access using .items() and unpack key/value
    print(disease)

    for patient in patients:
        print(patient)
```

```python
#-------------Dictionary containing dictionaries------------
patients = {
    "sam": {                    # <<--- here sam is key and value of it has dic... so dic will have key value pairs too
        "age": 50,
        "disease": "diabetic"
    },
    "tim": {
        "age": 60,
        "disease": "heart failure"
    }
}

for name, details in patients.items():
    print(name, details["age"], details["disease"])
```

```python
#-----------List of tuples containing lists--------------
data = [
    ("diabetic", ["Sam", "Kim"]),               # << -- This is tuple and within tuple , it has list
    ("heart failure", ["Tim", "Bim"])
]
for disease, patients in data:
    print(disease)

    for patient in patients:
        print(patient)
```
