```python
#----Complex for loop --------------

graph = {
    "Diagnosis" : ["Biopsy", "CT_Scan"],
    "Biopsy" : ["Diagnosis", "Treatment"]
}

result1 =  {key: len(detail_value) for key, detail_value in graph.items()}       # Creates a new dictionary using the node as the key and its connection count as the value.
result2 =  {key: (len(detail_value) * 3) for key, detail_value in graph.items()} 
print(result1)
print(result2)
```
