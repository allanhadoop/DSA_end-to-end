Balanced BST is a concept, while AVL Tree and Red-Black Tree are two actual implementations of that concept. There are two ways or methods to keey balanced tree ---

So how to create BST from below numbers - [10, 20, 30, 40, 50, 60, 70]
1. Middle is 40 so left values of 40 are [10, 20, 30] and right side is [50, 60, 70]
so we get

2. Middle of [10,20,30] is 20 and Middle of [50,60,70] is 60.

    Strict balance        Less strict balance
    using heights         using colors

                     ↓

                     BSTs

|                 | AVL                             | Red-Black                        |
| --------------- | ------------------------------- | -------------------------------- |
| BST?            | ✅                              | ✅                              |
| Self-balancing? | ✅                              | ✅                              |
| Balance         | More strict                     | Less strict                      |
| Uses            | Height/balance factor           | Red/black colors                 |
| Search          | Very good                       | Very good                        |
| Insert/Delete   | Can require more balancing      | Usually more flexible            |
| Main idea       | Keep tree more tightly balanced | Keep tree from becoming too tall |


```python
#-------------Balanced BST concept -------------
class Node:
    def __init__(self, value):
        self.value = value
        self.left = None
        self.right = None

def create_balanced_bst(values):
    if not values:
        return None
    middle = len(values) // 2                # For [10, 20, 30] . There are 3 values so 3 // 2 = 1 (Index) so it will be [10, 20, 30] -->20 which is at index 1
    root = Node(values[middle])               # Make the middle the root

    root.left = create_balanced_bst(values[:middle])            # <<--------Recursion
    root.right = create_balanced_bst(values[middle + 1:])       # <<--------Recursion
    return root

root = create_balanced_bst([10, 20, 30])

print(root.value)
print(root.left.value)
print(root.right.value)

"""
                [10,20,30]
                    |
             take middle = 20
                    |
                 root 20
                /      \
               /        \
           [10]          [30]
             ↓             ↓
          root 10       root 30
"""
```
