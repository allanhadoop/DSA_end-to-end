Balanced BST is a concept, while AVL Tree and Red-Black Tree are two actual implementations of that concept. There are two ways or methods to keey balanced tree ---

So how to create BST from below numbers - [10, 20, 30, 40, 50, 60, 70]

|                 | AVL                             | Red-Black                        |
| --------------- | ------------------------------- | -------------------------------- |
| BST?            | ✅                              | ✅                              |
| Self-balancing? | ✅                              | ✅                              |
| Balance         | More strict                     | Less strict                      |

| Main idea       | Keep tree more tightly balanced | Keep tree from becoming too tall |


```python
#-------------Balanced BST concept -------------
"""
                [10,20,30,40]
                       |
                       30  ← root
                     /  \
                   20    40
                  /
                10
"""
class Node:
    def __init__(self, value):
        self.value = value
        self.left = None
        self.right = None

def fun(values):
    if not values:
        return None
    middle = len(values) // 2                # For [10, 20, 30] . There are 4 values so 4 // 2 = 1 (Index) so it will be [10, 20, 30,40] -->30 which is at index 1
    root = Node(values[middle])              # Make the middle the root

    root.left = fun(values[:middle])            # <<--------Recursion       fun([10, 20])
    root.right = fun(values[middle + 1:])       # <<--------Recursion       fun([40])
    return root

root = fun(sorted([80, 20, 30,88 ]))        # << -- BST works correctly only when we sort the input 

print(root.value)
print(root.left.value)
print(root.right.value)
print(root.left.left.value)
```

```python
#---------AVL Tree Automatically balances (self-balancing tree)------------
# Build the tree → check how unbalanced it is → rotate → get a balanced tree.
"""
10
  \
   20
     \
      30
---Logic flow
Create tree
    ↓
    10
      \
       20
         \
          30
    ↓
Calculate heights
    ↓
10 = 3
20 = 2
30 = 1
    ↓
Calculate balance
    ↓
30 → 0
20 → -1
10 → -2
    ↓
-2 means too much RIGHT
    ↓
RR situation
    ↓
LEFT ROTATION
    ↓
       20
      /  \
    10    30
"""
class Node:
    def __init__(self, value):
        self.value = value
        self.left = None
        self.right = None
        self.height = 1                # << -- Assume every node starts with height = 1  so 10 → 20 → 30 even though isn't correct for the whole tree.

def height(node):
    if node is None:
        return 0
    return node.height

def balance_factor(node):                           # << -- For 30 , left = 0 and right = 0 so 0 - 0 = 0 balanced . For 20, 0 - 1 = -1 and for 10 0 -2 = -2 (Too much on the right)
    return height(node.left) - height(node.right)

def rotate_left(root):
    new_root = root.right                    # << -- root = 10, root.right = 20 so new_root = 20 so we will move it up
    root.right = new_root.left               # << - 20.right = 10
    new_root.left = root                     # << -- move 10 to left
    # Update heights
    root.height = 1 + max(height(root.left),
                          height(root.right))
    new_root.height = 1 + max(height(new_root.left),
                              height(new_root.right))
    print("Post rotation new height of 20 :", new_root.height)
    return new_root

# Create tree
root = Node(10)
root.right = Node(20)
root.right.right = Node(30)

root.right.height = 2
root.height = 3
# Calculate balance
print("Balance of 30:", balance_factor(root.right.right))
print("Balance of 20:", balance_factor(root.right))
print("Balance of 10:", balance_factor(root))

# Check if rotation is needed
if balance_factor(root) < -1:                     # << -- Check unbalanced criteria
    print("\nTree is too heavy on the RIGHT")
    print("RR case detected")
    print("Performing LEFT rotation...")
    root = rotate_left(root)

print("\nAfter rotation:")
print("Root:", root.value)
print("Left:", root.left.value)
print("Right:", root.right.value)
```

```python
# -- AVL where we will perform insertion and deletion and keep tree balanced at all times -- 
"""
1. insert values
2. automatically calculate height
3. automatically calculate balance factor
4. automatically rotate
5. delete values
6. automatically rebalance after deletion
"""

class Node:
    def __init__(self, value):
        self.value = value
        self.left = None
        self.right = None
        self.height = 1

def height(node):
    if node is None:
        return 0
    return node.height

def balance_factor(node):
    if node is None:
        return 0
    return height(node.left) - height(node.right)

def update_height(node):
    node.height = 1 + max(
        height(node.left),
        height(node.right)
    )

def rotate_right(root):
    new_root = root.left
    middle = new_root.right
    new_root.right = root
    root.left = middle
    update_height(root)
    update_height(new_root)
    return new_root

def rotate_left(root):
    new_root = root.right
    middle = new_root.left
    new_root.left = root
    root.right = middle
    update_height(root)
    update_height(new_root)
    return new_root

def rebalance(root):
    update_height(root)
    balance = balance_factor(root)
    # Too much LEFT
    if balance > 1:
        # LR case
        if balance_factor(root.left) < 0:
            root.left = rotate_left(root.left)
        # LL case
        return rotate_right(root)

    # Too much RIGHT
    if balance < -1:
        # RL case
        if balance_factor(root.right) > 0:
            root.right = rotate_right(root.right)
        # RR case
        return rotate_left(root)
    return root

def insert(root, value):
    # Empty tree
    if root is None:
        return Node(value)
    # Normal BST insertion
    if value < root.value:
        root.left = insert(root.left, value)
    elif value > root.value:
        root.right = insert(root.right, value)
    else:
        return root       # duplicate value

    # Balance the tree while coming back
    return rebalance(root)

def find_min(node):
    current = node
    while current.left is not None:
        current = current.left
    return current

def delete(root, value):
    if root is None:
        return None
    # Find the value
    if value < root.value:
        root.left = delete(root.left, value)
    elif value > root.value:
        root.right = delete(root.right, value)
    else:
        # Case 1: No child
        if root.left is None and root.right is None:
            return None

        # Case 2: Only right child
        elif root.left is None:
            return root.right
        # Case 3: Only left child
        elif root.right is None:
            return root.left

        # Case 4: Two children
        else:
            smallest = find_min(root.right)
            root.value = smallest.value
            root.right = delete(
                root.right,
                smallest.value
            )

    # Balance while coming back
    return rebalance(root)

def print_tree(root, level=0, position="Root"):
    if root is None:
        return
    print(
        "   " * level
        + position
        + ": "
        + str(root.value)
        + " (height="
        + str(root.height)
        + ", balance="
        + str(balance_factor(root))
        + ")"
    )
    print_tree(root.left, level + 1, "L")
    print_tree(root.right, level + 1, "R")

root = None
print("INSERT 10")
root = insert(root, 10)
print_tree(root)
print("\nINSERT 20")
root = insert(root, 20)
print_tree(root)

print("\nINSERT 30")
root = insert(root, 30)
print_tree(root)


print("\nINSERT 40")
root = insert(root, 40)
print_tree(root)


print("\nINSERT 50")
root = insert(root, 50)
print_tree(root)


print("\nDELETE 30")
root = delete(root, 30)
print_tree(root)


print("\nDELETE 40")
root = delete(root, 40)
print_tree(root)
```

```python
#------------Red black-----Automatically balances using color---Here each node gets a color 
""" 
       20 ⚫
      /     \
   10 🔴    30 🔴

AVL Thinking -- 
"How tall is my left side?
 How tall is my right side?"

Red-Black thinks:
"What colors are my nodes?
 Do my red/black rules still hold?"

                   BST
                   │
                   ↓
            Can become skewed
                   │
                   ↓
          Need to stay balanced
                   │
          ┌────────┴────────┐
          ↓                 ↓
        AVL              Red-Black
          │                 │
       Height              Colors
          │                 │
      Rotations       Recolor + Rotation
"""
class RBNode:

    def __init__(self, value, color):
        self.value = value
        self.color = color
        self.left = None
        self.right = None


root = RBNode(20, "BLACK")

root.left = RBNode(10, "RED")
root.right = RBNode(30, "RED")


print(root.value, root.color)
print(root.left.value, root.left.color)
print(root.right.value, root.right.color)
```
