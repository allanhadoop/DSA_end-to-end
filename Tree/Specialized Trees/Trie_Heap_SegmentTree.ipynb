```python
#--------------Simple Trie Program
"""
Support we have - ["cat", "car", "card", "care", "dog"]
After Trie it becomes - 
       root
      /    \
     c      d
     |      |
     a      o
    / \     |
   t   r    g         # #<< Here there is no left or right like BST tree.. It is actually stored as node as follows 
---

   a
├── t
└── r

a.children["t"]  → Node for t
a.children["r"]  → Node for r

"""
class Node:
    def __init__(self):
        self.children = {}     # << -- This stores the letters that can come next. {  "c": Node, "d": Node}
        self.end = False       # << -- Does a complete word end here?" when it becomes CAT then word completed then True

root = Node()
def insert(word):
    current = root
    for letter in word:
        if letter not in current.children:
            current.children[letter] = Node()
        current = current.children[letter]               # <<-- For example - 'c' becomes key 
    current.end = True

insert("cat")
insert("car")
insert("dog")
def search(word):
    current = root
    for letter in word:
        if letter not in current.children:
            return False
        current = current.children[letter]
    return current.end

print(search("cat"))
print(search("car"))
print(search("cow"))
```

```python
#-------Segment Tree - quickly answer questions about ranges.

"""
Position:    1   2   3   4   5   6   7
             ↓   ↓   ↓   ↓   ↓   ↓   ↓
numbers:     2   5   7   1   8   3   6

Index:       0   1   2   3   4   5   6
             ↓   ↓   ↓   ↓   ↓   ↓   ↓
numbers:     2   5   7   1   8   3   6

The tree will eventually look conceptually like:

                    [1-7]
                      32
                    /    \
                 [1-4]  [5-7]
                   15      17
                  /  \    /  \
               [1-2][3-4][5-6][7]
                 7    8    11   6
"""
numbers = [2, 5, 7, 1, 8, 3, 6]
tree = [0] * 20                    # This creates [0, 0, 0, 0, 0, 0, 0, 0, 0, ...] . We use this to store our tree. e.g tree[1], tree[2] to store the Segment Tree's information......
def build(node, start, end):                              # ---creates the Segment Tree.
    if start == end:
        tree[node] = numbers[start - 1]
        return
    middle = (start + end) // 2                           #  -- (1 + 7) // 2 = 4 so we split [1 -------- 7] into [1 --- 4] [5 --- 7]

    build(node * 2, start, middle)                        # << -- Build positions 1 to 4 and store the answer in tree[2].
    build(node * 2 + 1, middle + 1, end)                  # << -- Build right side
    tree[node] = tree[node * 2] + tree[node * 2 + 1]      # << -- # Add both children to answer question like What is the sum of positions 1 to 4?" So the Segment Tree is basically pre-calculating and storing answers for ranges.

def query(node, start, end, left, right):
    if right < start or end < left:                       # << --  Completely outside requested range
        return 0
    if left <= start and end <= right:                    # << --  Completely inside requested range
        return tree[node]
    middle = (start + end) // 2
    left_sum = query(node * 2, start, middle, left, right)
    right_sum = query(node * 2 + 1, middle + 1, end, left, right)
    return left_sum + right_sum

build(1, 1, 7)                                            # Build the tree for positions 1 through 7 because we have 7 numbers, and store the result in tree[1]. So node - 1, start 1 and end 7
answer = query(1, 1, 7, 2, 6)                             # Starting at tree node 1, whose range is 1–7, find the sum for requested range 2–6."
print(answer) 
```
