```python
#----------Tree DP(Dynamic Programming) -------
"""
              1
            /   \
           2     3
          / \     \
         4   5     6
            /
           7
Bottom nodes calculate first
    Nodes 4, 6 and 7
Their parents use those answers
    Nodes 5, 2 and 3
Root receives the final answer -------> Height = 4 
"""
class Node:
    def __init__(self, value):
        self.value = value
        self.left = None
        self.right = None

root = Node(1)
root.left = Node(2)
root.right = Node(3)
root.left.left = Node(4)
root.left.right = Node(5)
root.right.right = Node(6)
root.left.right.left = Node(7)

# Tree DP: calculate height
def height(node):
    if node is None:
        return 0
    left = height(node.left)
    right = height(node.right)
    return 1 + max(left, right)

print("Tree height:", height(root))
```

```python
#-----------Tree diameter ---------Longest path between two nodes ----------
"""
              1
            /   \
           2     3
          / \     \
         4   5     6
            /
           7

Longest path: 7 → 5 → 2 → 1 → 3 → 6
7 → 5 = 1 edge
5 → 2 = 1 edge
2 → 1 = 1 edge
1 → 3 = 1 edge
3 → 6 = 1 edge
----------------
Total   = 5 edges

"""
class Node:
    def __init__(self, value):
        self.value = value
        self.left = None
        self.right = None

root = Node(1)
root.left = Node(2)
root.right = Node(3)
root.left.left = Node(4)
root.left.right = Node(5)
root.right.right = Node(6)
root.left.right.left = Node(7)

def diameter(node):
    if node is None:
        return 0
    left_height = height(node.left)
    right_height = height(node.right)
    through_node = left_height + right_height
    left_diameter = diameter(node.left)
    right_diameter = diameter(node.right)
    return max(
        through_node,
        left_diameter,
        right_diameter
    )

def height(node):
    if node is None:
        return 0
    return 1 + max(height(node.left), height(node.right))

print("Tree diameter:", diameter(root))
```

```python
#---------Binary tree -- Binary Lifting is like saving shortcuts in advance. Instead of repeatedly travelling one step at a time, 
# you can jump 1, 2, 4, 8, 16 steps and combine those jumps to reach the ancestor you need.

parent = [None, None, 1, 2, 3, 4, 5, 6, 7, 8, 9]
# up[node][j] = ancestor 2**j steps above node
up = [[None] * 4 for _ in range(11)]
for node in range(2, 11):
    up[node][0] = parent[node]

for j in range(1, 4):
    for node in range(1, 11):
        previous = up[node][j - 1]
        if previous is not None:
            up[node][j] = up[previous][j - 1]

def ancestor(node, k):
    bit = 0
    while k > 0:
        if k % 2 == 1:
            node = up[node][bit]
            if node is None:
                return None
        k //= 2
        bit += 1
    return node

print(ancestor(10, 7))
```

```python
#-----------------Heavy-Light Decomposition
# HLD divides a tree into chains so that a long path can be handled as smaller, straight sections.
# Think of a tree as a road network. HLD groups selected roads into chains to make path calculations faster.

"""
              1
            /   \
           2     3
          / \     \
         4   5     6
            /
           7
Suppose you want to calculate the total cost of travelling from city 7 to city 4. Without HLD, you follow the road path: 7 → 5 → 2 → 4
Example - We want to travel from 7 to 4.
Step 1: HLD divides the tree into chains
Chain A
1 → 2 → 5 → 7
Chain B
2 → 4
Step 2: Start at person 7. Person 7 belongs to Chain A.
Follow the chain upward:
		7 — Start
		^
		5
		^
		2 — Common connection
We've reached person 2.
Step 3: Person 4 belongs to another chain. Person 4 is in Chain B, connected directly to person 2. So move down from person 2 to person 4.
		2
		v
		4 — Destination
Step 4: Combine the path
		7 → 5 → 2 → 4   ---- This is the complete path.
"""
class Node:
    def __init__(self, value):
        self.value = value
        self.children = []

# Create the tree
nodes = {i: Node(i) for i in range(1, 8)}
nodes[1].children = [nodes[2], nodes[3]]
nodes[2].children = [nodes[4], nodes[5]]
nodes[3].children = [nodes[6]]
nodes[5].children = [nodes[7]]

def calculate_size(node):
    size = 1
    for child in node.children:
        size += calculate_size(child)
    node.size = size
    return size

chains = []
# Choose the heavy child and form chains
def build_chains(node, chain):
    if not node.children:
        chains.append(chain)
        return
    heavy = max(node.children, key=lambda child: child.size)
    # Continue along the heavy chain
    build_chains(heavy, chain)
    # Start separate chains for the other children
    for child in node.children:
        if child != heavy:
            build_chains(child, [])

calculate_size(nodes[1])
build_chains(nodes[1], [])

for chain in chains:
    print(chain)
```

```python
#--------Tree seralization --- 
"""
Serialization
        Tree → String
        Save the tree as text.

Deserialization
        String → Tree
        Rebuild the original tree from text.
"""
class Node:
    def __init__(self, value):
        self.value = value
        self.left = None
        self.right = None

# Create the tree
root = Node(1)
root.left = Node(2)
root.right = Node(3)
root.left.left = Node(4)
root.left.right = Node(5)

def serialize(node):
    if node is None:
        return "None"
    return (
        str(node.value) + ","
        + serialize(node.left) + ","
        + serialize(node.right)
    )

saved_tree = serialize(root)

print(saved_tree)
def deserialize(data):
    values = iter(data.split(","))           # << --------- It turns the string into a list:
    def build():
        value = next(values)
        if value == "None":
            return None
        node = Node(int(value))
        node.left = build()
        node.right = build()
        return node
    return build()


new_root = deserialize(saved_tree)

def print_tree(node):
    if node is None:
        return
    print(node.value)
    print_tree(node.left)
    print_tree(node.right)

print_tree(new_root)
```
