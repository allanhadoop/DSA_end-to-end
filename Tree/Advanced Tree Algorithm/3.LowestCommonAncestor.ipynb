#--------Lowest Common Ancestor 
              10
             /  \
            5    20
           / \
          3   7

What is the lowest common ancestor of 3 and 7?        ------------>  5  (5 is the lowest node that contains both)

```python
#--------Find lowest Common Ancestor -- 
class treehead:
    def __init__(self, value):
        self.value = value
        self.left = None                 
        self.right = None               

root = treehead(10)
root.left = treehead(5)                  
root.right = treehead(20)                

root.left.left = treehead(3)             
root.left.right = treehead(7) 

def lowest_common_ancestor(root, p, q):
    if root is None:
        return None
    if root.value == p or root.value == q:
        return root
    left = lowest_common_ancestor(root.left, p, q)            # <<--- Recursive
    right = lowest_common_ancestor(root.right, p, q)          # <<--- Recursive

#---Below block of code ------------------------------
    if left is not None and right is not None:
        return root
    if left is not None:
        return left
    return right

"""
              Did left AND right find something?
                         |
                    YES → return root
                         |
                        NO
                         ↓
                 Did left find something?
                    /          \
                  YES           NO
                   ↓             ↓
              return left    return right
"""
answer = lowest_common_ancestor(root, 3, 7)
print(answer.value)
```

```python
#------------One more excercise ------------Lowest common ancestor --------
class treehead:
    def __init__(self, value):
        self.value = value
        self.left = None                 
        self.right = None               

root = treehead(0)
root.left = treehead(1.1)                # This has 2 childrens  
root.right = treehead(2)               # lonely node             

root.left.left = treehead(1.2)         # Child of 1.1   
root.left.right = treehead(1.3)        # Child of 1.1

def get_lowest_anc(root,p,q):
    if root is None:                                 # ----- Gowing down validation------------
        return None                                     #                  |
    if root.value == p or root.value == q:           #                  |
        return root                                  # ----------------------------------------

    value1 = get_lowest_anc(root.left,p,q)           #-- ----Recursive point for left-----------------
    value2 = get_lowest_anc(root.right,p,q)          #-- ----Recursive point for right-----------------

    if value1 is not None and value2 is not None:    # -------------While coming back validation---
        return root                                  #                  |
    if value1 is not None:                           #                  |
        return value1                                #                  |
    return value2                                    # ----------------------------------------

ancestor = get_lowest_anc(root,1.2,1.3)   
print(ancestor.value)
```
