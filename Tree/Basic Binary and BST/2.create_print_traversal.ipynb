```python
#----------Count All Nodes -- How many nodes are in my tree?
class one:
    def __init__(self, name):
        self.name = name
        self.children = []
H = one("Hospital")           #<<-- We are creating object and assigning name 
C = one("Cardiology")
O = one("Oncology")
I = one("ICU")
OP = one("OPD")
CH = one("Chemotherapy")

H.children = [C,O]              #<< -- Assigning 2 class objects to top node H inside list bracket that has names already in previous line
C.children = [I,OP]
O.children = [CH]

def count_nodes(one):
    count = 1                   # << -- Recursion approach -- Every function call gets its own count.
    for child in one.children:
        count = count + count_nodes(child)         # <<<-- First time it will have count_nodes(C) and count_nodes(O) and then within that their nodes. E.g. Cardiology returns 3 and oncology 2 to Hospital.
    return count
print(count_nodes(H))

"""
Hospital
count = 1
   │
   │ calls Cardiology
   ↓
Cardiology
count = 1
   │
   │ calls ICU
   ↓
ICU
count = 1
   │
   │ returns 1
   ↓
Cardiology
count = 1 + 1 = 2
   │
   │ calls OPD
   ↓
OPD
count = 1
   │
   │ returns 1
   ↓
Cardiology
count = 2 + 1 = 3
   │
   │ returns 3
   ↓
Hospital
count = 1 + 3 = 4
🔹 Every function call gets its own count.
🔹 The parent waits while the child runs.
🔹 The child returns its answer to the parent.
🔹 The parent adds that answer to its own count.
"""
```

```python
#----------Count leaf nodes -- leaf here in our examples are lowest level so we have 3 --     ICU     OPD      Chemotherapy 
class one:
    def __init__(self, name):
        self.name = name
        self.children = []
H = one("Hospital")           #<<-- We are creating object and assigning name 
C = one("Cardiology")
O = one("Oncology")
I = one("ICU")
OP = one("OPD")
CH = one("Chemotherapy")

H.children = [C,O]              #<< -- Assigning 2 class objects to top node H inside list bracket that has names already in previous line
C.children = [I,OP]
O.children = [CH]

def count_leaf_nodes(one):
    if len(one.children) == 0:          #<<-- This means lowest leaf reached so count as 1 into return
        return 1 
    total_leaf = 0
    for child in one.children:
        total_leaf = total_leaf + count_leaf_nodes(child)
    
    return total_leaf

print(count_leaf_nodes(H))
```

#------------Binary Tree ---------A binary tree allows maximum two children per node
              10
             /  \
            5    20
           / \   /
          3   7 15

```python
#------------Binary Tree(Going deeper into head side) ---------A binary tree allows maximum two children per node

class treehead:
    def __init__(self, value):
        self.value = value
        self.left = None                #<<-- Here left gets value from object definition in below code
        self.right = None               

root = treehead(10)

root.left = treehead(30)                 # <<--- Here we are now considering left as a class in itself and putting value to it
root.right = treehead(70)               # <<--- Here we are now considering right as a class in itself and putting value to it

root.left.left = treehead(20)            #<< --- Here we are recursively creating more class objects and adding value to it
root.left.right = treehead(40)

root.right.left = treehead(60)
root.right.right = treehead(80)

def print_all(treehead):                # <<--- Here we are taking class blueprint and not object
    if treehead is None:
        return 
    print(treehead.value)

    print_all(treehead.left)
    print_all(treehead.right)

print_all(root)                       # <<-- Here we are passing class object, not class itself
```

#----Binary Search Tree (BST) follows a rule: Left side  <  Root  <  Right side
Suppose we need to search for 60 .. so it is on right side of the tree so no point in search on the left side, So we go RIGHT.


              10
             /  \
           30    70
          /  \   / \
        20   40 60  80

```python
#----Binary Search Tree (BST) follows a rule: Left side  <  Root  <  Right side
class treehead:
    def __init__(self, value):
        self.value = value
        self.left = None                 
        self.right = None               

root = treehead(10)
root.left = treehead(30)                  
root.right = treehead(70)                

root.left.left = treehead(20)             
root.left.right = treehead(40)
root.right.left = treehead(60)
root.right.right = treehead(80)

def search(treehead, value):
    if treehead is None:
        return False
    if treehead.value == value:
        return True
    if value < treehead.value:
        return search(treehead.left, value)

    return search(treehead.right, value)

search(root, 80)
```

There are 3 traversal algorithms  to read tree as follows - 
1) Inorder -- Left → Print(Root) → Right
2) Preorder --Print(Root) → Left → Right
3) Postorder -- Left → Right → Print(Root)
              10
             /  \
            5    20
           / \
          3   7 
Inorder - 3 → 5 → 7 → 10 → 20
Preorder -- 10 → 5 → 3 → 7 → 20
Postorder - 3 → 7 → 5 → 20 → 10

Pre-order → Root comes before children
In-order → Root comes in between left and right
Post-order → Root comes after children

```python
#------Tree Traversal----Visit every node in a particular order.
class treehead:
    def __init__(self, value):
        self.value = value
        self.left = None                 
        self.right = None               

root = treehead(10)
root.left = treehead(5)                  
root.right = treehead(20)                

root.left.left = treehead(3)             
root.left.right = treehead(7) 

#--------In-order- (so print(root) in between)
def inorder(treehead):         ## <<-- Here we are taking only class as blueprint argument and not class.class
    if treehead is None:
        return
    inorder(treehead.left)
    print(treehead.value)
    inorder(treehead.right)

inorder(root.left)             ## <<-- Here we are passing object.object to the function
print(root.value)
inorder(root.right)

print("-------------------------------------")
#--------Post-order - (print(root) will come at the end)
def postorder(treehead):         ## <<-- Here we are taking only class as blueprint argument and not class.class
    if treehead is None:
        return
    postorder(treehead.left)
    postorder(treehead.right)
    print(treehead.value)

postorder(root.left)             ## <<-- Here we are passing object.object to the function
postorder(root.right)
print(root.value)
```

FIFO = First In, First Out
Queue -- The first person who enters the line is the first person served or taken out and then next.

deque (Double-ended queue) -- is a  data structure/class from the collections module.It allows us to efficiently add/remove items from either end.
from collections import deque

queue.append(...) -- to add something to the right/end.
queue.popleft()   -- to remove something from the left/front.

                 deque
       ┌────────────────────┐
LEFT   │  10  20  30  40    │ RIGHT
       └────────────────────┘
         ↑              ↑
      popleft()       append()                   # <<--- These are methods 

      

```python
#----------Level Order / BFS -- Instead of going deep first, we can process the tree level by level and print the levels.
# "add at the back, take from the front" behavior is what makes the BFS work.

from collections import deque
class treehead:
    def __init__(self, value):
        self.value = value
        self.left = None                 
        self.right = None               

root = treehead(10)
root.left = treehead(5)                  
root.right = treehead(20)                

root.left.left = treehead(3)             
root.left.right = treehead(7) 

def level_order(treehead):
    queue = deque([treehead])             # -- if treehead = 10 then queue = [10] . queue as a waiting line.
    while queue:
        node = queue.popleft()            # Take the first person from left from the line so before [10] and now it will be [] so node = 10
        print(node.value)                 # Print 10.
        if node.left:                     # Put 10's children into the queue which are 5 and 20
            queue.append(node.left)       # add 5 
        if node.right:
            queue.append(node.right)      # add 20    so now queue waiting line will be [5, 20] and goes to "while" loop above

level_order(root)

""" Below is flow of level_order ---
Tree:

        10
       /  \
      5    20
     / \
    3   7


START
Queue = [10]

↓ popleft()

Print 10
Queue = []

Add 5, 20
Queue = [5, 20]

↓ popleft()

Print 5
Queue = [20]

Add 3, 7
Queue = [20, 3, 7]

↓ popleft()

Print 20
Queue = [3, 7]

↓ popleft()

Print 3
Queue = [7]

↓ popleft()

Print 7
Queue = []

STOP
"""
```

#--- Find Tree Height --- How many levels/depth does the tree have?

              10        Height = 1(for self) + How many childrens height (2  below line) so 1 + max(2 for 5 and 1 self for 20) which is 2 so total = 3  
             /  \
            5    20     Height = 1(for self) + childrens height (1 based on below line) so total height at this level = 1 + max(1 for 3,1 for 20) = 2 
           / \
          3   7         Height = 1(for self) + how many childrens max (0 for 3,0 for 7) = 0 so height = 1 + 0 = 1

so Height = 3 
For every node, ask:
"What is the height of my left side?"
"What is the height of my right side?"
"Which one is bigger?"

Then add 1 for myself.
My height = 1 + height of my tallest side
return 1 + max(left_height, right_height)

----------------
              10
             /  \
            5    20
           / \
          3   7
3 → height = 1
7 → height = 1

5 → 1 + max(1,1)
   → 2                  # - Return value is not lost . It is stored here inside the height(5) call:

20 → height = 1         # - Return value is not lost . It is stored here inside the height(20) call:

10 → 1 + max(2,1)      
    → 3                 # - This is final return value but earlier return values are not lost or disappeared so no need to use any variable to store it

```python
#--- Find Tree Height --- How many levels/depth does the tree have?
class treehead:
    def __init__(self, value):
        self.value = value
        self.left = None                 
        self.right = None               

root = treehead("top")
root.left = treehead("l1_left")                  
root.right = treehead("l1-right-alone")                

root.left.left = treehead("l1-l2-left")             
root.left.right = treehead("l1-l2-right") 

def find_height(treehead):
    if treehead is None:          # << -- if node is None for e.g. f1-f2-left.left is: none then : return 0 = "There is no node here, stop recursion."
        return 0
    Left_H = find_height(treehead.left)
    Right_H = find_height(treehead.right)
    return (1 + max(Left_H, Right_H))   # << -- 1 for self and then add max of left or right height and return value is not lost.

find_height(root)
"""
TOP:
"I need L1's answer."
        ↓
L1:
"I need L2's answer."
        ↓
L2:
"I can answer!"
        ↓
        1
        ↑
L1:
"Great! Now I can answer."
        ↑
        2
        ↑
TOP:
"Great! Now I can answer."
        ↑
        3
"""
```
