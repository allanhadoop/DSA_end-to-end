```python
def count(n):
    if n == 0:
        return
    print("Going down:", n)

    count(n - 1)              # <<-- Calling same function again here which will work only till it going down but this line doesnot takes count value of previously called function while coming up
    print("Coming back:", n)

count(3)
```
