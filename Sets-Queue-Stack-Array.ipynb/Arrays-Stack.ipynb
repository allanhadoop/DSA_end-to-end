```python
#--- Rotate an array to the right
"""
Before: [1, 2, 3, 4, 5]
After:  [4, 5, 1, 2, 3]
"""
numbers = [1, 2, 3, 4, 5]
k = 2

k = k % len(numbers)            # << - The % operation handles cases where the rotation count is larger than the array length.
numbers[:] = numbers[-k:] + numbers[:-k]
print(numbers)
```

```python
#--- Kadane's algorithm -- Find the maximum sum of a continuous section

numbers = [-2, 1, -3, 4, -1, 2, 1, -5, 4]
"""
4 + (-1) + 2 + 1 = 6
Maximum sum = 6
Why is this the best section? Because its sum, 6, is greater than the sum of any other continuous section in this list.
"""

current_sum = numbers[0]
best_sum = numbers[0]

for number in numbers[1:]:
    current_sum = max(number, current_sum + number)
    best_sum = max(best_sum, current_sum)

print(best_sum)
"""
current_sum = max(number, current_sum + number)
- number means start a new section with just this number.
- current_sum + number means continue the existing section.
- max() chooses whichever gives the larger sum.

Number	Decision	            Current sum
4	    Start a new section  	    4
-1	    Continue: 4 - 1	            3
2	    Continue: 3 + 2	            5
1	    Continue: 5 + 1	            6
-5	    Continuing gives 1, 
        better to restart at -5	    -5
4	    Start fresh at 4	        4

The program remembers the best sum it has seen, which is 6.
"""
```

```python
#----------Merge overlapping intervals - This is useful for combining overlapping calendar events or time ranges. 
"""
intervals = [[1, 3], [2, 6], [8, 10], [9, 12]]
        [1, 3] and [2, 6] overlap, so merge them into [1, 6].
        [8, 10] does not overlap with [1, 6], so start a new interval.
        [9, 12] overlaps with [8, 10], so merge them into [8, 12].
"""
intervals = [[1, 3], [2, 6], [8, 10], [9, 12]]
intervals.sort()
merged = []
for start, end in intervals:
    if not merged or start > merged[-1][1]:
        merged.append([start, end])
    else:
        merged[-1][1] = max(merged[-1][1], end)
print(merged)
```

```python
#--------------A stack follows LIFO — Last In
# Program 1 — Validate a medical report
# Problem: A healthcare system receives structured test data. Check whether its brackets are correctly opened and closed.
"""
Imagine the system reads:
{ "tests": [ "CBC", "LFT" ] }
1. It sees { and puts it on the stack.
2. It sees [ and puts that on top.
3. It sees ] and removes [.
4. It sees } and removes {.
5. The stack is empty, so the brackets are balanced.
"""
text = '{"tests": ["CBC", "LFT"]}'
stack = []
pairs = {")": "(", "]": "[", "}": "{"}
for char in text:
    if char in "([{":
        stack.append(char)

    elif char in ")]}":
        if not stack or stack.pop() != pairs[char]:
            print("Invalid report structure")
            break
else:
    print("Valid structure" if len(stack) == 0
          else "Invalid report structure")
```

```python
#----------Stack--- Find the next higher lab result
# Problem: A lab system records a sequence of illustrative readings for one patient. 
# For each reading, find the first higher reading to its right.
"""
8 → next higher reading is 11.
6 → next higher reading is 11.
11 → no higher reading follows it.
9 → no higher reading follows it.
"""
readings = [8, 6, 11, 9]

result = [-1] * len(readings)
stack = []

for i, reading in enumerate(readings):
    while stack and readings[stack[-1]] < reading:
        index = stack.pop()
        result[index] = reading
    stack.append(i)

print(result)
```

```python
#---Stack --- Find the largest rectangle in a hospital workload histogram
# Problem: Suppose a hospital tracks the number of patients requiring attention during six consecutive time slots.
"""
Counts:  [2, 1, 5, 6, 2, 3]
                 ┌───┐
             ┌───│   │
             │ 5 │ 6 │
             └───┴───┘
The bars with heights 5 and 6 can form a rectangle with:
- Height = 5
- Width = 2 consecutive slots
- Area = 5 × 2 = 10
"""
patients = [2, 1, 5, 6, 2, 3]

stack = []
max_area = 0

for i in range(len(patients) + 1):
    current = patients[i] if i < len(patients) else 0

    while stack and patients[stack[-1]] > current:
        height = patients[stack.pop()]
        left = stack[-1] if stack else -1
        width = i - left - 1

        max_area = max(max_area, height * width)

    stack.append(i)

print(max_area)
```

```python
#--------Evaluate a medical calculation using a stack 
# Problem: A clinical software component needs to evaluate an arithmetic 
# expression after its inputs and formula have been validated.
tokens = ["2", "3", "+", "4", "*"]
stack = []

for token in tokens:
    if token.isdigit():
        stack.append(int(token))

    else:
        b = stack.pop()
        a = stack.pop()

        if token == "+":
            stack.append(a + b)
        elif token == "-":
            stack.append(a - b)
        elif token == "*":
            stack.append(a * b)
        elif token == "/":
            stack.append(a // b)

print(stack.pop())
"""
Token	        Action	                    Stack
2	            Push 2	                    [2]
3	            Push 3	                    [2, 3]
+	            Pop 2 and 3; add them	    [5]
4	            Push 4	                    [5, 4]
*	            Pop 5 and 4; multiply them	[20]
"""
```
