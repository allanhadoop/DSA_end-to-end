```python
#--------Queue-----------A queue follows FIFO — First In, First Out.
# Program 1 — Hard: Generate binary numbers using a queue
# Generate binary representations of numbers from 1 to 5.
# 1 10 11 100 101
""" 
Start with "1" in the queue.
Number removed	            Add to the queue
1	                        10, 11
10	                        100, 101
11	                        110, 111
100	                        1000, 1001
101	                        1010, 1011
The queue processes numbers in order, producing the binary sequence.
"""
from collections import deque

queue = deque(["1"])

for _ in range(5):
    number = queue.popleft()
    print(number, end=" ")

    queue.append(number + "0")
    queue.append(number + "1")
```

```python
#-------------Find the first non-repeating character in a stream
# Suppose characters arrive one at a time: a, b, a, c
# After each arrival, find the first character that has appeared only once so far.
# The queue keeps characters in arrival order. Repeated characters at the front are removed until the first 
# unique character is found.
from collections import deque, Counter

characters = "abac"
queue = deque()
counts = Counter()

for char in characters:
    counts[char] += 1
    queue.append(char)

    while queue and counts[queue[0]] > 1:
        queue.popleft()

    print(queue[0] if queue else "-")
```

```python
#------ Find the maximum in every sliding window"
"""
Numbers: [1, 3, -1, -3, 5, 3, 6, 7]
Window size: 3
Find the maximum number in each group of three consecutive numbers.
Window	            Maximum
[1, 3, -1]	            3
[3, -1, -3]	            3
[-1, -3, 5]	            5
[-3, 5, 3]	            5
[5, 3, 6]	            6
[3, 6, 7]	            7
The queue stores indices, not just values. It removes expired indices and values that cannot become future maximums.
"""
from collections import deque

numbers = [1, 3, -1, -3, 5, 3, 6, 7]
k = 3
queue = deque()
result = []
for i, number in enumerate(numbers):
    while queue and queue[0] <= i - k:
        queue.popleft()
    while queue and numbers[queue[-1]] <= number:
        queue.pop()
    queue.append(i)

    if i >= k - 1:
        result.append(numbers[queue[0]])
print(result)
```

```python
#--------Implement a circular queue
#  A circular queue reuses empty spaces instead of continually shifting elements.
"""
How it works
1. Fill the queue with 10, 20, and 30.
2. Remove 10, freeing its position.
3. Insert 40, which reuses that position.
The % operator allows the position to wrap around to the beginning. In a production implementation, 
you would also handle display order separately from the underlying storage order.
"""
class CircularQueue:
    def __init__(self, size):
        self.data = [None] * size
        self.size = size
        self.front = 0
        self.count = 0
    def enqueue(self, value):
        if self.count == self.size:
            print("Queue full")
            return
        rear = (self.front + self.count) % self.size
        self.data[rear] = value
        self.count += 1
    def dequeue(self):
        if self.count == 0:
            print("Queue empty")
            return

        value = self.data[self.front]
        self.front = (self.front + 1) % self.size
        self.count -= 1
        return value

q = CircularQueue(3)
q.enqueue(10)
q.enqueue(20)
q.enqueue(30)

print(q.dequeue())
q.enqueue(40)
print(q.data)
```

```python
#--Sets -- Find whether a pair adds up to a target
# Given [4, 8, 1, 6], determine whether two different values add up to 10.
"""
See 4; remember it.
See 8; remember it.
See 1; remember it.
See 6; the needed value is 10 - 6 = 4.
4 is already in the set, so the answer is True.
"""
numbers = [4, 8, 1, 6]
target = 10
seen = set()
found = False

for number in numbers:
    if target - number in seen:
        found = True
        break
    seen.add(number)

print(found)
```

```python
#------Sets -- Find the longest consecutive sequence
#  [100, 4, 200, 1, 3, 2] Find the longest sequence of consecutive values, regardless of their original order.
"""
Start a sequence only when the previous number is absent.
For example, 1 is a starting point because 0 is absent.
Check for 2, then 3, then 4.
The sequence length is 4.
"""
numbers = [100, 4, 200, 1, 3, 2]
values = set(numbers)
longest = 0

for number in values:
    if number - 1 not in values:
        length = 1

        while number + length in values:
            length += 1

        longest = max(longest, length)

print(longest)
```

```python
#----sets - Find the longest substring without repeated characters
#  "abcabcbb"  The longest substring with no repeated characters is "abc", with length 3.
"""
a, b, c are all unique, so the window length is 3.
The next a is a duplicate, so move the left boundary until the old a is removed.
Continue moving the window while keeping its characters unique.
The maximum length remains 3.
"""
text = "abcabcbb"
seen = set()
left = 0
longest = 0

for right in range(len(text)):
    while text[right] in seen:
        seen.remove(text[left])
        left += 1

    seen.add(text[right])
    longest = max(longest, right - left + 1)

print(longest)
```
