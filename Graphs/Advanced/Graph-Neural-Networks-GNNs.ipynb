```python
#-------Graph Neural Networks (GNNs) - A normal machine-learning model looks at patient information. A GNN can also learn from how that information is connected.
"""
A simplified GNN lets each item gather information from its neighboring items.
        Step 1: The mutation receives information
        The EGFR mutation is connected to Osimertinib, so it can gather information from the drug.
        Mutation's own value: 10 · Neighbor's value: 20

        Step 2: The drug receives information
        The drug is connected to both the mutation and the documented response.
        Drug's own value: 20 · Neighbor values: 10 and 30

        Step 3: The response receives information
        The response gathers information from the connected drug.
        Response's own value: 30 · Neighbor's value: 20

This process is called message passing: each node gathers information from its neighbors and combines it with its own information.
---
Difference between a normal ML model and a GNN
----------------------------------------------
Normal machine learning	                        Graph Neural Network
-------------------------                       -------------------------------------------
Looks at patient features	                    Looks at features and connections
Example: mutation, age, lab results	            Example: mutation connected to drug and outcome
Learns patterns from the input features	        Can also learn patterns from neighboring nodes
----------Below program----
            The EGFR mutation received information from the connected drug.
            The drug received information from both the mutation and the documented response.
            The response received information from the drug.
            Each node combined its original value with the average of its neighbors' values.
"""
# Simplified oncology knowledge graph. # This is an illustration of GNN message passing,

graph = {
    "EGFR_Mutation"         : ["Osimertinib"],
    "Osimertinib"           : ["EGFR_Mutation", "Tumor_Response"],
    "Tumor_Response"        : ["Osimertinib"]
}
# Each node starts with a simple numeric feature. # These values are illustrative, not clinical measurements.
features = {
    "EGFR_Mutation": 1.0,
    "Osimertinib": 0.5,
    "Tumor_Response": 1.0
}
new_features = {}
for node, neighbors in graph.items():
    neighbor_values = [                                         # Gather information from connected nodes.
        features[neighbor] for neighbor in neighbors
    ]
    neighbor_average = (
        sum(neighbor_values) / len(neighbor_values)
        if neighbor_values else 0
    )
    new_features[node] = (                                      # Combine the node's own information with its neighbors.
        features[node] + neighbor_average
    ) / 2
print("Original node features:")
print(features)

print("\nFeatures after one message-passing step:")
print(new_features)
```
