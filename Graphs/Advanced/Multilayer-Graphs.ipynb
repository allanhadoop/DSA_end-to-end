```python
#----Multilayer graphs -- How can we connect different kinds of oncology information without mixing them all into one undifferentiated network?
"""
A cancer patient has different kinds of medical information:
- Diagnosis: Lung cancer
- Biopsy: Tissue examination report
- Molecular test: EGFR mutation detected
- Treatment: Osimertinib recorded
If we put everything into one big network without distinguishing the information types, it can become difficult to 
understand which record belongs to which category.
A multilayer graph solves this by keeping different types of information in separate layers while connecting related information.

"""
# Four separate layers of oncology information.
layers = {"Clinical"    : {"Diagnosis": ["Biopsy"]},
          "Pathology"   : {"Biopsy": ["Pathology_Report"]},
          "Molecular"   : {"EGFR_Result": ["EGFR_L858R"]},
          "Treatment"   : {"Therapy": ["Osimertinib"]}
        }
cross_layer_links = [                       # Links connecting different layers.
         ("Diagnosis", "Biopsy"),
         ("Pathology_Report", "EGFR_Result"),
         ("EGFR_L858R", "Therapy"),
         ("Therapy", "Osimertinib")
        ]
print("Information layers:")
for layer, relationships in layers.items():
    print(f"\n{layer} layer:")
    for source, targets in relationships.items():
        print(f"  {source} -> {targets}")
print("\nCross-layer relationships:")
for source, target in cross_layer_links:
    print(f"{source} -> {target}")
```
