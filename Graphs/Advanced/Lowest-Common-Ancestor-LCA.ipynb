```python
#----Lowest Common Ancestor (LCA)---Find the closest shared event between two branches of a clinical history.
# Move both events up the tree until they meet. That meeting point is their lowest common ancestor.
"""
For example, a patient has a biopsy, then separate biomarker and imaging branches. What is the nearest shared ancestor of 
Targeted_Therapy and Progression_Review?
"""
tree = {
    "Diagnosis"             : ["Biopsy"],
    "Biopsy"                : ["Biomarker_Test", "CT_Scan"],
    "Biomarker_Test"        : ["Targeted_Therapy"],
    "Targeted_Therapy"      : [],
    "CT_Scan"               : ["Progression_Review"],
    "Progression_Review"    : []
}
parent = {"Diagnosis": None}
depth = {"Diagnosis": 0}

def build_tree(node):
    for child in tree[node]:
        parent[child] = node
        depth[child] = depth[node] + 1
        build_tree(child)

def lca(a, b):
    while depth[a] > depth[b]:
        a = parent[a]
    while depth[b] > depth[a]:
        b = parent[b]
    while a != b:
        a = parent[a]
        b = parent[b]
    return a
build_tree("Diagnosis")
print(lca("Targeted_Therapy", "Progression_Review"))
```
