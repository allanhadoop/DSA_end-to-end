```python
#--- Graph Reversal -- 
"""
Suppose a patient has this oncology history:
Original graph — forward clinical timeline
Diagnosis-------->Biomarker test---->Treatment--->Clinical outcome: Tumor response

Reversed graph — trace backward from the outcome
Tumor response--->Treatment--->Biomarker test--->Diagnosis

Conceptual illustration: reversing edges changes traversal direction, not the underlying clinical facts.
-------
For example, a reviewer asks:
“This patient showed a tumor response. Which treatment, biomarker test, and diagnosis are linked to that outcome?”
Instead of searching forward from the diagnosis, we reverse the edges and start at the outcome.
"""

graph = {
    "Diagnosis"             : ["Biomarker_Test"],
    "Biomarker_Test"        : ["Treatment"],
    "Treatment"             : ["Tumor_Response"],
    "Tumor_Response"        : []
}

def reverse_graph(graph):
    reversed_graph = {}
    for node in graph:                     #  ----Keep every original node.
        reversed_graph[node] = []
    for source in graph:                   # -----Reverse every directed edge.
        for destination in graph[source]:
            reversed_graph[destination].append(source)
    return reversed_graph

reversed_graph = reverse_graph(graph)
print("Reversed graph:")
for node, neighbors in reversed_graph.items():
    print(node, "->", neighbors)

def trace_backward(graph, starting_node):     # Start from the outcome and trace backward.
    visited = set()
    stack = [starting_node]
    while stack:
        current = stack.pop()
        if current not in visited:
            visited.add(current)
            print(current)
            for neighbor in graph[current]:
                if neighbor not in visited:
                    stack.append(neighbor)
print("\nTrace backward from Tumor_Response:")
trace_backward(reversed_graph, "Tumor_Response")
```
