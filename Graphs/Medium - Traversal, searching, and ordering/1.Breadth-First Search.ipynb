```python
#---------BFS (Breadth-First Search) -- Explore nearby connections first, level by level.
# - Oncology use case: Starting from a lung cancer diagnosis, find all related clinical records within two relationship steps.
# -- BFS finds records by increasing distance from the starting record. 
# This can help retrieve nearby linked records in an oncology history.
# -- Time complexity: \(O(V+E)\), where \(V\) is the number of nodes and \(E\) is the number of edges.

from collections import deque
graph = {
    "Diagnosis"       : ["Biopsy", "EGFR_Test"],
    "Biopsy"          : ["Diagnosis", "Pathology_Report"],
    "EGFR_Test"       : ["Diagnosis", "Molecular_Report"],
    "Pathology_Report": ["Biopsy", "Tumor_Board"],
    "Molecular_Report": ["EGFR_Test", "Tumor_Board"],
    "Tumor_Board"     : ["Pathology_Report", "Molecular_Report",
                        "Treatment_Plan"],
    "Treatment_Plan"  : ["Tumor_Board"]
}

def bfs_within_steps(graph, start, max_steps):
    queue = deque([(start, 0)])                    # --queue stores records waiting to be explored.
    visited = {start}
    while queue:
        node, steps = queue.popleft()              # - popleft() removes the oldest record first.
        print(f"{node}: {steps} step(s) from diagnosis")
        if steps == max_steps:                     # --steps tracks the number of relationships from the diagnosis.
            continue
        for neighbor in graph[node]:
            if neighbor not in visited:
                visited.add(neighbor)
                queue.append((neighbor, steps + 1))
bfs_within_steps(graph, "Diagnosis", 2)
```
