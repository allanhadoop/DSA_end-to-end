```python
#---- Depth-First Search-- Meaning: Follow one path as deeply as possible before exploring another path.
# -- Oncology use case: Trace a patient's connected history from diagnosis through biopsy, pathology, tumor board, and treatment.
"""
    Start at Diagnosis.
    Visit Biopsy, the first neighbor in the list.
    Continue to Pathology_Report, then Tumor_Board.
    Continue down an available path until no unvisited neighbors remain.
    Backtrack and explore any remaining branches.
"""

graph = {
    "Diagnosis"             : ["Biopsy", "EGFR_Test"],                 # <<-- From Biopsys it goes to Pathology report
    "Biopsy"                : ["Diagnosis", "Pathology_Report"],
    "EGFR_Test"             : ["Diagnosis", "Molecular_Report"],
    "Pathology_Report"      : ["Biopsy", "Tumor_Board"],                # << -- Here we have biopsy. Then we take tumor board from here 
    "Molecular_Report"      : ["EGFR_Test", "Tumor_Board"],      
    "Tumor_Board"           : ["Pathology_Report", "Molecular_Report",   #<< -- Here we came from Pathology report tumor board value
                                "Treatment_Plan"],
    "Treatment_Plan"        : ["Tumor_Board"]
}

def dfs(graph, node, visited):
    visited.add(node)
    print("Visited:", node)

    for neighbor in graph[node]:
        if neighbor not in visited:
            dfs(graph, neighbor, visited)

dfs(graph, "Diagnosis", set())
```
