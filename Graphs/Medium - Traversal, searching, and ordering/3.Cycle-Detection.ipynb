```python
#---- Cycle Detection--- dentify a loop in a graph.
# -- Oncology use case: Detect circular references in linked clinical records, such as a pathology record linking 
# to a tumor board record that incorrectly links back to an earlier workflow step. For this example, we use a directed graph to make the loop explicit.
"""
Biopsy
Pathology Report
Tumor Board
Back to Biopsy —> cycle

Key learning: Cycle detection can flag circular references or workflow dependencies that need investigation. 
However, cycles are not always errors in a clinical knowledge graph; a patient's records may legitimately refer to one another.

Time complexity: \(O(V+E)\).
"""
graph = {
    "Diagnosis": ["Biopsy"],
    "Biopsy": ["Pathology_Report"],
    "Pathology_Report": ["Tumor_Board"],
    "Tumor_Board": ["Biopsy"],  # Incorrect circular link
    "Treatment_Plan": []
}

def has_cycle(graph):
    visited = set()                  # <<-- here we use set -- visited contains every node already explored.
    active_path = set()              # << -- active_path contains nodes on the current DFS path.
    def dfs(node):
        visited.add(node)
        active_path.add(node)
        for neighbor in graph[node]:
            if neighbor not in visited:
                if dfs(neighbor):
                    return True
            elif neighbor in active_path:         # << -- If an edge points to a node already in active_path, a cycle exists.
                return True
        active_path.remove(node)
        return False

    for node in graph:
        if node not in visited:
            if dfs(node):
                return True
    return False

if has_cycle(graph):
    print("Cycle detected: review clinical record links")
else:
    print("No cycle detected")

```
