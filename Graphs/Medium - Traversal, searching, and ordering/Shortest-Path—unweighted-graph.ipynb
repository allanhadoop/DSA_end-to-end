```python
#-----------Shortest path — unweighted graph - Find a path with the fewest edges (connections), without considering weights or costs.
#  Suppose an oncologist wants to trace the shortest chain of records connecting a lung cancer diagnosis to a treatment outcome.
#  Each connection counts as one edge. BFS (Breadth-First Search) explores the nearest nodes first and finds a path 
#  with the fewest edges in an unweighted graph.
"""
BFS finds the path with the fewest connections, not necessarily the most clinically meaningful or chronologically correct path. 
A real oncology history graph should also validate timestamps and relationships
"""

from collections import deque
# Each key is a clinical record.
# Each value is a list of directly connected records.
graph = {
    "Diagnosis"             : ["Biopsy", "Clinical_Summary"],
    "Biopsy"                : ["EGFR_Test"],
    "EGFR_Test"             : ["Treatment"],
    "Treatment"             : ["Outcome"],
    "Clinical_Summary"      : ["Outcome"],
    "Outcome"               : []
}

def shortest_path(graph, start, target):
    queue = deque([(start, [start])])
    visited = {start}
    while queue:
        current, path = queue.popleft()
        if current == target:
            return path
        for neighbor in graph[current]:
            if neighbor not in visited:
                visited.add(neighbor)
                queue.append((neighbor, path + [neighbor]))
    return None

path = shortest_path(graph, "Diagnosis", "Outcome")
print("Shortest path:", path)
print("Number of edges:", len(path) - 1)
```
