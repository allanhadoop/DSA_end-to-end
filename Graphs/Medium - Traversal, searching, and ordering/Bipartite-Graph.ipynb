```python
#-------Bipartite graph --- Divide nodes into two separate groups. Connections go from one group to the other, not between nodes in the same group.
# -- A cancer registry wants to identify which patients have which diagnoses.
"""
The two groups are:
- Group 1: Patients
- Group 2: Diagnoses
        Group 1: Patients
        Patient_101
        Patient_102
        Patient_103

        Group 2: Diagnoses
        Lung cancer
        Breast cancer


Connections represent patient–diagnosis relationships. A patient can connect to multiple diagnoses, and a diagnosis can 
connect to multiple patients.

        For example:
        - Patient 101 → Lung cancer
        - Patient 102 → Lung cancer
        - Patient 102 → Breast cancer
        - Patient 103 → Breast cancer
        
The bipartite structure is a modeling choice, not a requirement that each patient have only one diagnosis. In a real system, 
diagnosis dates, status, evidence, and source records should also be represented.
Another useful application is a patient–clinical-trial graph, where the connections indicate potential eligibility for review, 
not confirmed eligibility or enrollment.
"""

patients = [                    # --------Group 1 of Node
    "Patient_101",
    "Patient_102",
    "Patient_103"
]
diagnoses = [                   # --------Group 2 of Node
    "Lung_Cancer",
    "Breast_Cancer"
]
# Connections between patients and diagnoses
edges = [
    ("Patient_101", "Lung_Cancer"),
    ("Patient_102", "Lung_Cancer"),
    ("Patient_102", "Breast_Cancer"),
    ("Patient_103", "Breast_Cancer")
]
# Build adjacency lists for both groups
graph = {}
for patient in patients:
    graph[patient] = []
for diagnosis in diagnoses:
    graph[diagnosis] = []
for patient, diagnosis in edges:
    graph[patient].append(diagnosis)
    graph[diagnosis].append(patient)

# Find patients associated with lung cancer
target = "Lung_Cancer"
print("Patients with", target + ":")
for patient in graph[target]:
    print(patient)
# Find diagnoses associated with Patient_102
print("\nDiagnoses for Patient_102:")
for diagnosis in graph["Patient_102"]:
    print(diagnosis)
```
