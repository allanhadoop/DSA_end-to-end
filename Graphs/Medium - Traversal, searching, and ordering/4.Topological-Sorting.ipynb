```python
#-------Topological Sorting -- Arrange nodes in an order that respects their dependencies.
"""
Oncology use case: Order a simplified clinical workflow so that a biopsy precedes pathology review, 
molecular results are available before the tumor board decision, and the treatment plan follows the review.

Unlike the previous programs, this algorithm requires a directed acyclic graph (DAG) for a complete ordering.
--
    indegree = {node: 0 for node in graph}

    is represented as follows - 
            indegree = {
            "Diagnosis": 0,
            "Biopsy": 0,
            "Treatment": 0
        }
    see the output of program -- after biopsy --> EGFR test -->Pathology ---> then Molecular_Report
"""

from collections import deque

graph = {
    "Diagnosis"             : ["Biopsy", "EGFR_Test"],
    "Biopsy"                : ["Pathology_Report"],
    "EGFR_Test"             : ["Molecular_Report"],
    "Pathology_Report"      : ["Tumor_Board"],
    "Molecular_Report"      : ["Tumor_Board"],
    "Tumor_Board"           : ["Treatment_Plan"],
    "Treatment_Plan"        : []
}

def topological_sort(graph):
    indegree = {node: 0 for node in graph}         # -- Set all incoming connection to 0 for all graph keys
    # Count incoming dependencies for each node
    for node in graph:
        for neighbor in graph[node]:
            indegree[neighbor] += 1
    # Begin with nodes that have no prerequisites
    queue = deque(
        node for node in graph if indegree[node] == 0
    )
    order = []
    while queue:
        node = queue.popleft()
        order.append(node)
        for neighbor in graph[node]:
            indegree[neighbor] -= 1
            if indegree[neighbor] == 0:
                queue.append(neighbor)
    if len(order) != len(graph):
        print("Cannot order workflow: cycle detected")
    else:
        print("Valid dependency order:")
        for step, node in enumerate(order, start=1):
            print(f"{step}. {node}")
topological_sort(graph)
```
