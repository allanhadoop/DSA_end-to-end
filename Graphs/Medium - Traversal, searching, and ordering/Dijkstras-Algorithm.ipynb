```python
#-- Dijkstra's algorithm - Find the path with the lowest total cost when all edge weights are non-negative.
# - Suppose a patient needs an oncology referral. Different healthcare facilities have different referral travel times. 
# Dijkstra's algorithm finds the route with the lowest total travel time.
"""
From	        To	            Travel time
-----           -----           ----------
Hospital A  	Hospital B	    20 minutes
Hospital A	    Hospital C	    10 minutes
Hospital C	    Hospital B	    5 minutes

BFS treats every edge as one step. Dijkstra considers the numerical weight of every edge. Dijkstra can optimize travel time, 
financial cost, or another non-negative measure. In practice, hospital referrals should also consider specialist availability, 
required services, and urgency; shortest travel time alone does not determine the appropriate referral.

"""
import heapq
# Each connection has a travel-time weight.
graph = {
    "Hospital_A"        : [("Hospital_B", 20), ("Hospital_C", 10)],
    "Hospital_B"        : [("Cancer_Center_D", 15)],
    "Hospital_C"        : [("Hospital_B", 5),("Cancer_Center_D", 40)],
    "Cancer_Center_D"   : []
}
def dijkstra(graph, start, target):
    
    priority_queue = [(0, start, [start])]                   #---STRUCTURE is follows --> (total cost, current node, path)
    best_cost = {start: 0}
    while priority_queue:
        cost, current, path = heapq.heappop(priority_queue)  #--take out the item with the smallest priority value.
        if cost > best_cost.get(current, float("inf")):
            continue
        if current == target:
            return cost, path
        for neighbor, weight in graph[current]:
            new_cost = cost + weight
            if new_cost < best_cost.get(neighbor, float("inf")):
                best_cost[neighbor] = new_cost
                heapq.heappush(priority_queue,(new_cost, neighbor, path + [neighbor]))  #---heappush() means: add an item to the priority queue. E.g. (10, "Hospital_C", ["Hospital_A", "Hospital_C"])
    return float("inf"), None           # -- Inf means infinity . It runs when the algorithm has searched all reachable routes but has not found the target hospital.
"""
float("inf") represents an infinite cost, and None represents the absence of a path. 
Together, they tell the calling code that no route was found.
"""


cost, path = dijkstra(graph, "Hospital_A", "Cancer_Center_D")
print("Best route:", path)
print("Total travel time:", cost, "minutes")
```
