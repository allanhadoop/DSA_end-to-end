```python
#--- Connected-component detection — first identify the two groups, then investigate whether they should be connected. 
# We do not automatically connect them.The value is not simply connecting everything. It is identifying disconnected clinical 
# records and helping determine whether important relationships are missing.
# For example, if a patient's treatment response is not linked to the treatment they received, the oncology history may be 
# incomplete. Detecting this gap could help reduce manual chart review.


# Find separate groups of oncology records. # Each connection represents a relationship between records.
edges = [
    ("Diagnosis", "Biopsy"),
    ("Biopsy", "Treatment"),
    ("CT_Scan", "Tumor_Response"),
    ("Mutation_Test", "Targeted_Therapy"),
]
graph = {}                                        # - # Create an undirected graph for connectivity detection.
for record1, record2 in edges:
    graph.setdefault(record1, []).append(record2) #--If record1 is not already in the dictionary, create an empty list for it. Then add record2 to that list.
    graph.setdefault(record2, []).append(record1) #- setdefault() = get or create the list.

def find_connected_components(graph):
    visited = set()
    components = []

    for record in graph:                          # Check every record, including disconnected records.
        if record not in visited:
            component = []
            stack = [record]
            visited.add(record)
            while stack:                          # Explore all connected records using DFS.
                current = stack.pop()
                component.append(current)
                for neighbor in graph[current]:
                    if neighbor not in visited:
                        visited.add(neighbor)
                        stack.append(neighbor)
            components.append(component)
    return components

components = find_connected_components(graph)
print("Connected components:")
for number, component in enumerate(components, start=1):
    print(f"Group {number}: {component}")
```
