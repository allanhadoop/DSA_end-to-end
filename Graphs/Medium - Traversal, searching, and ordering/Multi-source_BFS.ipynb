```python
#----Multi-source BFS--Start BFS from multiple nodes at the same time to find the nearest nodes or records to any of those starting points.
#-- Healthcare use case
"""
Suppose an oncology patient has several known clinical events:
    - An EGFR mutation test
    - A treatment start
    - A toxicity event
You want to identify other records that are closest to any of these events in the graph.
For this example, “closest” means the fewest graph edges, not the shortest time interval.The result tells us that Outcome is 
two edges away from the nearest source record. It does not tell us which source is nearest unless we also track the source 
responsible for each discovery. This can help with graph-based record retrieval, such as gathering clinical records near several 
known events. It does not automatically reconstruct the correct clinical timeline: temporal ordering and source-document 
verification must be handled separately.
"""

from collections import deque
graph = {
    "EGFR_Test"             : ["Mutation_Result"],
    "Mutation_Result"       : ["EGFR_Test", "Treatment_Record"],
    "Treatment_Record"      : ["Mutation_Result","Treatment_Start","Toxicity_Event"],
    "Treatment_Start"       : ["Treatment_Record", "Follow_Up"],
    "Toxicity_Event"        : ["Treatment_Record", "Lab_Result"],
    "Follow_Up"             : ["Treatment_Start", "Outcome"],
    "Lab_Result"            : ["Toxicity_Event"],
    "Outcome"               : ["Follow_Up"]
}
def multi_source_bfs(graph, sources):
    queue = deque()
    distance = {}
    for source in sources:            #-- # Add all starting records before searching
        if source not in distance:
            distance[source] = 0
            queue.append(source)
    while queue:
        current = queue.popleft()
        for neighbor in graph[current]:
            if neighbor not in distance:
                distance[neighbor] = distance[current] + 1
                queue.append(neighbor)
    return distance

sources = [ "EGFR_Test","Treatment_Start","Toxicity_Event"]
distances = multi_source_bfs(graph, sources)
print("Distance from the nearest source record:")
for record, distance in distances.items():
    print(record, "->", distance, "edge(s)")
```
