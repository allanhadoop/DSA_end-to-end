```python
#-------------Cycle -- A cycle is a sequence of connections that returns to its starting node.
"""
Oncology use case
Suppose a clinical knowledge graph contains these relationships:
Drug_A → Drug_B → Drug_C → Drug_A
This creates a circular reference. A system may need to flag this for review, depending on what the relationships mean.

This example uses depth-first search (DFS) to explore the graph.
- visited stores nodes already explored.
- active stores nodes on the current exploration path.
- If the program reaches a node already in active, it has found a cycle.
- If it finishes exploring a node without finding a cycle, it removes that node from active.
"""
graph = {
    "Drug_A": ["Drug_B"],
    "Drug_B": ["Drug_C"],
    "Drug_C": ["Drug_A"],
    "Drug_D": []
}
def has_cycle(graph):
    visited = set()
    active = set()
    def visit(node):
        if node in active:
            return True
        if node in visited:
            return False
        visited.add(node)
        active.add(node)

        for neighbor in graph[node]:
            if visit(neighbor):
                return True
        active.remove(node)
        return False

    for node in graph:
        if visit(node):
            return True
    return False

if has_cycle(graph):
    print("Cycle detected")
else:
    print("No cycle detected")
```
