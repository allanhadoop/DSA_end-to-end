```python
# -- Adjacency list -- For example, if you ask, “Which record is connected directly to the patient's 
# record?”, the adjacency list lets you look it up.
# Space: \(O(V+E)\)
# Look up a node's neighbors: \(O(1)\) to access the list, plus \(O(k)\) to process its \(k\) neighbors.

graph = { 
    "P101": ["D101"],                #---Dictionary
    "D101": ["T101"],
    "T101": ["TR101"],
    "TR101": []
}
# Find records directly connected to the patient
patient_id = "P101"

print("Records directly connected to", patient_id)
for record in graph[patient_id]:
    print(record)
```

```python
#-----Adjacency matrix — Check whether two records are connected
# --- Create a table where 1 means a connection exists and 0 means no connection exists.
"""
matrix[0][1] = 1: Patient P101 is directly connected to diagnosis D101.
matrix[1][2] = 1: Diagnosis D101 is directly connected to test T101.

Space: \(O(V^2)\), because the matrix stores a cell for every pair of nodes.
Check a connection: \(O(1)\) after the node indexes are known.
"""
nodes = ["P101", "D101", "T101", "TR101"]
matrix = [                                       # 1 = direct connection, 0 = no direct connection
    [0, 1, 0, 0],  # P101
    [0, 0, 1, 0],  # D101
    [0, 0, 0, 1],  # T101
    [0, 0, 0, 0]   # TR101
]

# Check if patient is directly connected to diagnosis
patient_index = nodes.index("P101")
diagnosis_index = nodes.index("D101")
if matrix[patient_index][diagnosis_index] == 1:
    print("Patient is directly connected to diagnosis")
else:
    print("No direct connection found")
# Check if patient is directly connected to treatment
treatment_index = nodes.index("TR101")
if matrix[patient_index][treatment_index] == 1:
    print("Patient is directly connected to treatment")
else:
    print("No direct connection to treatment")
```
