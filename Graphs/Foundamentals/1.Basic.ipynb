```python

# Healthcare graph: nodes and connections
# Storing V nodes and E edges takes O(V+E) space in this representation.

nodes = [ "P101 - Patient","D101 - Lung cancer", "T101 - EGFR test", "TR101 - Treatment record"]
edges = [ ("P101", "D101"), ("D101", "T101"), ("T101", "TR101")]

print("Healthcare nodes:")
for node in nodes:
    print(node)

print("\nHealthcare connections:")
for start, end in edges:                    # ---Each pair (start, end) describes a relationship.
    print(start, "->", end)  
```
