```python
#-------Path--- A path is a sequence of nodes connected in order.
"""
Oncology use case
Trace a fictional patient journey:   Diagnosis → Biopsy → Treatment → Follow-up

"""

graph = {
    "Diagnosis": ["Biopsy"],
    "Biopsy": ["Treatment"],
    "Treatment": ["Follow_Up"],
    "Follow_Up": []
}
path = [
    "Diagnosis",
    "Biopsy",
    "Treatment",
    "Follow_Up"
]

# Check whether each consecutive pair is connected
valid_path = True
for i in range(len(path) - 1):
    current_node = path[i]
    next_node = path[i + 1]
    if next_node not in graph[current_node]:
        valid_path = False
        break

if valid_path:
    print("Valid path:", " -> ".join(path))
else:
    print("Invalid path")
```
