```python
#-------Directed graph — Connections have a direction
# --A relationship goes from one record to another in a specified direction. Diagnosis → Treatment record
# The arrow indicates the direction of the relationship. It does not automatically mean the diagnosis caused the treatment; 
# it simply represents the relationship we chose to record.
"""
For an adjacency-list representation, storing the graph takes \(O(V+E)\) space. 
Following one node's outgoing connections takes \(O(k)\), where \(k\) is the number of outgoing edges.
"""

graph = {
    "P101": ["D101"],
    "D101": ["T101"],
    "T101": ["TR101"],
    "TR101": []
}
# Follow the direction from diagnosis
start = "D101"

print("Starting from:", start)
for next_record in graph[start]:
    print(start, "->", next_record)
# Check the reverse direction
if start in graph["TR101"]:
    print("Treatment points back to diagnosis")
else:
    print("Treatment does not directly point to diagnosis")
```
