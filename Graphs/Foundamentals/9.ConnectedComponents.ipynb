```python
#--------Connected components -- A connected component is a group of nodes connected to one another, with no connections to other groups.
"""
Oncology use case
Suppose a dataset contains two groups of oncology records:
- Group 1: Patient 1, Diagnosis 1, Treatment 1
- Group 2: Patient 2, Diagnosis 2
If no graph connections exist between the two groups, they form two separate connected components.

---
Connected components can help identify disconnected groups of records, incomplete relationships, or isolated knowledge-graph sections. 
However, disconnected records are not necessarily data errors; they may simply belong to separate patients or cases.
"""

graph = {
    "Patient_1": ["Diagnosis_1"],
    "Diagnosis_1": ["Patient_1", "Treatment_1"],
    "Treatment_1": ["Diagnosis_1"],

    "Patient_2": ["Diagnosis_2"],
    "Diagnosis_2": ["Patient_2"]
}
def find_components(graph):
    visited = set()
    components = []
    for node in graph:
        if node not in visited:
            stack = [node]
            component = []
            while stack:
                current = stack.pop()
                if current not in visited:
                    visited.add(current)
                    component.append(current)
                    stack.extend(graph[current])
            components.append(component)
    return components

components = find_components(graph)
for i, component in enumerate(components, start=1):
    print("Component", i, ":", component)
print("Total components:", len(components))
```
