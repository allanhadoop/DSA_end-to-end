```python
#----------Weighted graph---Each connection has a value, such as time, cost, distance, or risk score.
"""
Oncology use case
Record the number of days between fictional treatment journey events:
- Diagnosis → Biopsy: 2 days
- Biopsy → Tumor Board: 3 days
- Tumor Board → Treatment Start: 5 days
graph = {
    key : [(e1,e2), (e1,e2).... ]    # graph is a dictionary whose values are lists containing tuples of elements.
}

A weighted graph stores additional information on each connection. Here, the weights represent elapsed time, 
not a clinical risk or treatment recommendation.
"""

graph = {
    "Diagnosis": [("Biopsy", 2)],                 # graph is a dictionary whose values are lists containing tuples of elements.
    "Biopsy": [("Tumor_Board", 3)],
    "Tumor_Board": [("Treatment_Start", 5)],
    "Treatment_Start": []
}

# Display the time between each event
for event, connections in graph.items():
    for next_event, days in connections:
        print(event, "->", next_event, ":", days, "days")
```
