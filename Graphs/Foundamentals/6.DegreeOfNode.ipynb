```python
#----------Degree of a node -- The degree of a node is the number of connections it has.
"""
Oncology use case
Suppose a Lung_Cancer diagnosis is associated with three investigations:
- CT Scan
- Biopsy
- Molecular Testing
We want to count how many investigations are directly connected to the diagnosis.
"""
graph = {
    "Lung_Cancer": ["CT_Scan", "Biopsy", "Molecular_Test"],
    "CT_Scan": ["Lung_Cancer"],
    "Biopsy": ["Lung_Cancer"],
    "Molecular_Test": ["Lung_Cancer"]
}

node = "Lung_Cancer"                    #-- FInd degree for a given node value
degree = len(graph[node])
print("Node:", node)
print("Number of connections:", degree)
```
