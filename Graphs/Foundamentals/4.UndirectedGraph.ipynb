```python
# --------Undirected graph--A connection works in both directions: A ↔ B.
# Suppose a clinical knowledge graph associates two findings: EGFR_Mutation and Lung_Cancer. 
# For this example, we represent their association as undirected, meaning either node can be used to reach the other.
"""
graph is a Python dictionary.
Each key represents a node.
Each list contains the nodes connected to that key.
We store the relationship in both directions.
"""

graph = {                             # -- Dictionary with key and value as list
    "EGFR_Mutation": ["Lung_Cancer"],               
    "Lung_Cancer": ["EGFR_Mutation"]
}

# Check connection in both directions
print(graph["EGFR_Mutation"])
print(graph["Lung_Cancer"])
```
