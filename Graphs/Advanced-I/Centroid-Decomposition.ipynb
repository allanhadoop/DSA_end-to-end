```python
#-----Centroid Decomposition-- Divide a large tree-shaped oncology referral network into smaller groups for distance queries.
"""
Hospital A ---> Hospital B ---> Hospital C ---> Hospital D 
A centroid is a node that splits a tree into smaller parts, with no remaining part containing more than half of the original nodes.
Let's choose Hospital B as the centroid.

If we remove B, the network splits into:
- Group 1: Hospital A — 1 hospital
- Group 2: Hospital C — Hospital D — 2 hospitals

There are four hospitals originally. Half of four is two, so neither group has more than two hospitals. 
Therefore, B is a valid centroid.
"""
graph = {
    "Hospital_A": ["Hospital_B", "Hospital_C"],
    "Hospital_B": ["Hospital_A", "Hospital_D", "Hospital_E"],
    "Hospital_C": ["Hospital_A", "Hospital_F", "Hospital_G"],
    "Hospital_D": ["Hospital_B"],
    "Hospital_E": ["Hospital_B"],
    "Hospital_F": ["Hospital_C"],
    "Hospital_G": ["Hospital_C"]
}
removed = set()
centroid_parent = {}

def find_centroid(start):
    size = {}
    def calculate(node, parent):
        size[node] = 1
        for neighbor in graph[node]:
            if neighbor != parent and neighbor not in removed:
                size[node] += calculate(neighbor, node)
        return size[node]
    total = calculate(start, None)
    def locate(node, parent):
        for neighbor in graph[node]:
            if neighbor != parent and neighbor not in removed:
                if size[neighbor] > total // 2:
                    return locate(neighbor, node)
        return node
    return locate(start, None)
def decompose(start, parent_centroid=None):
    centroid = find_centroid(start)
    centroid_parent[centroid] = parent_centroid
    removed.add(centroid)
    for neighbor in graph[centroid]:
        if neighbor not in removed:
            decompose(neighbor, centroid)

decompose("Hospital_A")
print("Centroid hierarchy:")
for hospital, parent in centroid_parent.items():
    print(hospital, "<-", parent)
```
