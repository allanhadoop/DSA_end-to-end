```python
#--------Graph embeddings-------Can we turn clinical information into numbers so a computer can search for potentially similar records?
"""
A graph contains nodes and relationships. A graph embedding represents a node or relationship as a numerical vector, such as:

EGFR_L858R → [0.8, 0.2, 0.9]

These numbers are not manually assigned clinical meanings. In a real embedding model, they are learned from data and 
a training objective.For this introductory example, we'll use illustrative vectors to demonstrate how similarity 
search works. We will not train an actual embedding model.

The program calculates the cosine similarity between the query vector and each other vector. A score closer to 1 means 
the vectors point in a more similar direction.

For these illustrative vectors, Case_B and Case_D should rank above Case_C in similarity to Case_A.
Healthcare value: A properly trained and validated embedding system could help retrieve potentially relevant clinical cases, 
related medical documents, or similar sections of an oncology history for human review.
But there is an important limitation: similarity is not clinical equivalence. Two cases may look similar numerically 
while differing in disease stage, mutations, prior therapies, or treatment outcomes. Real systems must validate 
retrieval quality and account for clinically important differences.

"""
import math
# Illustrative vectors, not real clinical embeddings.
# Each entity is represented by three numbers.
embeddings = {
    "Case_A": [0.9, 0.8, 0.2],
    "Case_B": [0.8, 0.9, 0.3],
    "Case_C": [0.1, 0.2, 0.9],
    "Case_D": [0.7, 0.8, 0.2]
}
def cosine_similarity(vector_a, vector_b):
    dot_product = sum(
        a * b for a, b in zip(vector_a, vector_b)
    )
    magnitude_a = math.sqrt(sum(a * a for a in vector_a))
    magnitude_b = math.sqrt(sum(b * b for b in vector_b))
    if magnitude_a == 0 or magnitude_b == 0:
        return 0
    return dot_product / (magnitude_a * magnitude_b)

query_case = "Case_A"
similarities = []
for case, vector in embeddings.items():
    if case == query_case:
        continue
    similarity = cosine_similarity(
        embeddings[query_case], vector
    )
    similarities.append((case, similarity))
similarities.sort(key=lambda item: item[1], reverse=True)
print("Cases most similar to", query_case)
for case, similarity in similarities:
    print(f"{case}: {similarity:.3f}")
```
