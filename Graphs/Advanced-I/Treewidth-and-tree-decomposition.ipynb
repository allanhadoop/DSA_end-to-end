```python
#-------Treewidth and tree decomposition--------Tree decomposition means grouping connected activities into smaller, overlapping groups called bags.
"""
Imagine a hospital workflow with three connected activities:
- A: Biomarker review
- B: Oncology review
- C: Treatment planning

A: Biomarker review -->B: Oncology review --> C: Treatment planning.
For our example, we can create two bags:
        Bag 1
        Biomarker review (A)
        Oncology review (B)

        Shared activity: B                 <------

        Bag 2
        Oncology review (B)
        Treatment planning (C)

Oncology review (B) appears in both bags. Think of bags as small working groups. The shared activity lets the groups fit together without losing the connection.
"""
#------------
"""
Treewidth is a number that helps describe how complicated a graph is in terms of its tree decomposition.
The formula is:  Treewidth = {Largest bag size}-1

In our example:
Bag	            Number of activities
---             -------------------
Bag 1	        2
Bag 2	        2
The largest bag contains 2 activities.  Treewidth}=2-1= boxed{1}
So, the treewidth is 1.

1) Tree decomposition: Divides a complicated graph into overlapping bags.
2) Treewidth: Measures the size of the largest bag in the best valid decomposition, minus one.
"""
# Original workflow graph.
workflow = {
    "Biomarker_Review"  : ["Oncology_Review"],
    "Oncology_Review"   : ["Biomarker_Review","Treatment_Planning"],
    "Treatment_Planning": ["Oncology_Review"]
}
bags = {                                                # Each bag contains a subset of original graph nodes.
    "Bag_1": {"Biomarker_Review","Oncology_Review"},
    "Bag_2": {"Oncology_Review","Treatment_Planning"}
}
bag_tree = {
    "Bag_1": ["Bag_2"],
    "Bag_2": ["Bag_1"]
}
print("Tree decomposition bags:")
for bag, nodes in bags.items():
    print(bag, ":", sorted(nodes))
all_bag_nodes = set().union(*bags.values())             # Check whether every original node appears in a bag.
all_nodes_covered = all(
    node in all_bag_nodes for node in workflow
)
all_edges_covered = True                                # Check whether every original edge appears together in some bag.
for node, neighbors in workflow.items():
    for neighbor in neighbors:
        together = any(                                 # "any" will return True if at least one bag passes the check
            node in bag and neighbor in bag
            for bag in bags.values()
        )
        if not together:
            all_edges_covered = False
print("\nAll nodes covered:", all_nodes_covered)
print("All edges covered:", all_edges_covered)
bags_with_oncology = [                                  # Check the running-intersection property for Oncology_Review.
    bag for bag, nodes in bags.items()
    if "Oncology_Review" in nodes
]
print(
    "Bags containing Oncology_Review:",
    bags_with_oncology
)
```
