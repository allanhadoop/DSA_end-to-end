```python
#-----Subgraph isomorphism----Can we find a particular pattern of clinical events inside a much larger oncology knowledge graph?
"""
Suppose researchers want to find cases containing this pattern:

>>>   Biomarker Test → Diagnosis → Treatment

In real data, the pattern might be more detailed, such as a documented biomarker result followed by a treatment 
decision and a subsequent outcome.

Potential oncology use: Identify recurring sequences of documented investigations, diagnoses, treatments, 
and outcomes for retrospective research.
--------Python code defines these connections:---
Example 1: Lung cancer
EGFR_Test
Lung_Diagnosis
Osimertinib

Example 2: Breast cancer
HER2_Test
Breast_Diagnosis
Targeted_Therapy

If all three checks pass, the program saves the matching pattern. The graph also contains other connections, such as CT_Scan → Lung_Diagnosis. That connection does not start with a 
biomarker test, so it doesn't match the required pattern.
"""
# A larger oncology event graph. # Arrows represent documented relationships.

graph = {
    "EGFR_Test"         : ["Lung_Diagnosis"],
    "Lung_Diagnosis"    : ["Osimertinib"],
    "Osimertinib"       : ["Response_Assessment"],
    "HER2_Test"         : ["Breast_Diagnosis"],
    "Breast_Diagnosis"  : ["Targeted_Therapy"],
    "Targeted_Therapy"  : ["Response_Assessment_2"],
    "CT_Scan"           : ["Lung_Diagnosis"]
}
# The pattern we want to find: # Biomarker test -> Diagnosis -> Treatment
node_types = {
    "EGFR_Test"             : "Biomarker",
    "HER2_Test"             : "Biomarker",
    "Lung_Diagnosis"        : "Diagnosis",
    "Breast_Diagnosis"      : "Diagnosis",
    "Osimertinib"           : "Treatment",
    "Targeted_Therapy"      : "Treatment",
    "Response_Assessment"   : "Outcome",
    "Response_Assessment_2" : "Outcome",
    "CT_Scan"               : "Investigation"
}
def find_patterns(graph, node_types):
    matches = []
    for start, neighbors in graph.items():
        if node_types[start] != "Biomarker":
            continue
        for diagnosis in neighbors:
            if node_types[diagnosis] != "Diagnosis":
                continue
            for treatment in graph[diagnosis]:
                if node_types[treatment] == "Treatment":
                    matches.append(
                        (start, diagnosis, treatment)
                    )
    return matches
matches = find_patterns(graph, node_types)
print("Matching clinical patterns:")
for match in matches:
    print(" -> ".join(match))
print("Total patterns found:", len(matches))
```
