```python
#----Graph algorithms on streaming data---How can we update a patient's oncology history as new clinical events arrive, 
# without rebuilding the entire graph every time?

"""
Imagine that a patient's history is initially:
Diagnosis → Biopsy
Later, the system receives new events:
- A biomarker result
- A treatment start
- A follow-up assessment
A streaming approach processes each new event and adds the appropriate node and connection incrementally.
"""
# Initial oncology patient-history graph.
graph = {
    "Diagnosis" : ["Biopsy"],
    "Biopsy"    : []
}
def add_event(graph, previous_event, new_event):
    if new_event not in graph:                          # Create the new event if it doesn't exist.
        graph[new_event] = []
    if previous_event not in graph:                     # Create the previous event if it doesn't exist.
        graph[previous_event] = []
    if new_event not in graph[previous_event]:          # Add the relationship, avoiding duplicates.
        graph[previous_event].append(new_event)
    print(f"Added: {previous_event} -> {new_event}")

incoming_events = [                                     # Events arrive over time.
    ("Biopsy",          "EGFR_Result"),
    ("EGFR_Result",     "Treatment_Start"),
    ("Treatment_Start", "Response_Assessment")
]
for previous_event, new_event in incoming_events:
    add_event(graph, previous_event, new_event)
print("\nUpdated oncology history:")
for event, connections in graph.items():
    print(event, "->", connections)
```
