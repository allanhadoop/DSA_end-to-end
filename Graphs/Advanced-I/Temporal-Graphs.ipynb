```python
#----Temporal graphs -- How can we represent not only which medical events are connected, but also when they occurred?
"""
Consider this illustrative patient history:
- January 5     : Diagnosis
- January 12    : Biopsy
- January 18    : Biomarker test
- January 25    : Treatment starts
- March 25      : Response assessment
A normal graph shows the relationships. A temporal graph also records when the events happened or when a relationship was valid.
For a real oncology history, distinguish the date an event occurred, the date it was documented, and the date the system 
learned about it. A report entered in March might describe a test performed in January.

"""
from datetime import datetime
events = {                                                                      #----------- Each event has a date and a type.
    "Diagnosis"             : {"date": "2026-01-05","type": "Clinical"},      # Dictionaries
    "Biopsy"                : {"date": "2026-01-12","type": "Pathology"},
    "Biomarker_Test"        : {"date": "2026-01-18","type": "Molecular"},
    "Treatment"             : {"date": "2026-01-25","type": "Treatment"},
    "Response_Assessment"   : {"date": "2026-03-25","type": "Outcome"}
}
edges = [                                       # Relationships between events.
    ("Diagnosis", "Biopsy"),
    ("Biopsy", "Biomarker_Test"),
    ("Biomarker_Test", "Treatment"),
    ("Treatment", "Response_Assessment")
]
def event_date(name):
    return datetime.strptime(
        events[name]["date"], "%Y-%m-%d"
    )
timeline = sorted(events, key=event_date)       # Sort events chronologically.
print("Oncology timeline:")
for name in timeline:
    print(events[name]["date"], ":", name)
print("\nDays between connected events:")
for source, target in edges:
    days = (event_date(target) - event_date(source)).days
    print(f"{source} -> {target}: {days} days")
```
