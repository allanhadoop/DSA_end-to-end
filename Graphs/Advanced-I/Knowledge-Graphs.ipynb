```python
#--------Knowledge graphs---Can we represent medical entities and their relationships in a form that a program can search and explain?
"""
A knowledge graph uses meaningful relationships such as:
- Patient → has diagnosis → Lung cancer
- Patient → has biomarker → EGFR L858R
- Patient → received treatment → Osimertinib
- Treatment record → documented in → Hospital report
Unlike a simple list, a knowledge graph tells us how the pieces of information relate to one another.

A program can start with a patient and find associated diagnoses, biomarkers, treatments, and source documents. 
It can also navigate from a particular biomarker back to the patient and the report that documented it.

For example, if a clinician asks, "Where did this biomarker result come from?", the system can 
follow the DOCUMENTED_IN relationship.For your oncology-history product, this is one of the most relevant concepts. 
However, the graph alone is not the differentiator. The value comes from accurately extracting events from documents, 
preserving provenance, representing time and uncertainty, and making it easy to verify every important relationship.
"""
# Each relationship is:
# (source entity, relationship, target entity)
knowledge_graph = [
    ("Patient_101", "HAS_DIAGNOSIS",        "Lung_Cancer"),
    ("Patient_101", "HAS_BIOMARKER",        "EGFR_L858R"),
    ("Patient_101", "RECEIVED_TREATMENT",   "Osimertinib"),
    ("EGFR_L858R",  "DOCUMENTED_IN",        "Molecular_Report_01"),
    ("Osimertinib", "DOCUMENTED_IN",        "Treatment_Record_02"),
    ("Lung_Cancer", "DOCUMENTED_IN",        "Discharge_Summary_03")
]
def find_relationships(entity):
    print(f"\nRelationships for {entity}:")
    for source, relationship, target in knowledge_graph:
        if source == entity:
            print(f"  {relationship} -> {target}")
        elif target == entity:
            print(f"  <- {relationship} <- {source}")
find_relationships("Patient_101")
find_relationships("EGFR_L858R")
```
