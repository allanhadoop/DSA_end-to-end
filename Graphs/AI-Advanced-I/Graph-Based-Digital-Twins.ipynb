```python
#------Graph-Based Digital Twins -- Maintain an evolving model of a real-world system, updating its state as new events are 
# recorded. A digital twin is more than a static graph. It represents the state of a system and how that state changes over time.

"""
A simplified patient-history representation might evolve as follows:
- Day 1: Diagnosis recorded.
- Day 5: Biomarker result recorded.
- Day 10: Treatment started.
- Day 40: Follow-up response assessment recorded.
The system maintains both the event history and the latest documented state.
The program preserves the history of events and maintains a current-state summary.
For example, the latest documented treatment is Osimertinib, while the history still preserves the original diagnosis and 
biomarker result. However, this is a minimal digital-twin prototype, not a clinically complete digital twin. It does not 
simulate tumor biology, predict future disease progression, or infer treatment effectiveness.
"""

from datetime import date
patient_twin = {                    #--- A simplified oncology history model.
    "patient_id": "STUDY_001",
    "events": [],
    "current_state": {
        "diagnosis": None,
        "biomarker": None,
        "treatment": None,
        "response" : None
    }
}
def record_event(event_date, event_type, value, source):
    event = {
        "date"  : event_date,
        "type"  : event_type,
        "value" : value,
        "source": source
    }
    patient_twin["events"].append(event)                    # Preserve the event in the longitudinal history.

    if event_type in patient_twin["current_state"]:         # Update the latest recorded state.
        patient_twin["current_state"][event_type] = value
    print(f"Recorded {event_type} on {event_date}")

record_event(
    "2026-01-01",
    "diagnosis",
    "Lung cancer",
    "Pathology_Report_001"
)
record_event(
    "2026-01-05",
    "biomarker",
    "EGFR mutation documented",
    "Molecular_Report_001"
)
record_event(
    "2026-01-10",
    "treatment",
    "Osimertinib started",
    "Treatment_Record_001"
)
record_event(
    "2026-02-10",
    "response",
    "Assessment documented; interpretation pending",
    "Imaging_Report_001"
)
print("\nLatest documented state:")
for key, value in patient_twin["current_state"].items():
    print(f"{key}: {value}")

print("\nTotal recorded events:", len(patient_twin["events"]))
```
