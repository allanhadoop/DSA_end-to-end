```python
#----Graph Diffusion Models --as an AI that starts with a messy molecular structure and gradually improves it until 
# it creates a possible new molecule. Imagine that AI has learned from thousands of existing molecular structures.
"""
The key idea is that the model learns how to reverse corruption:
1. During training, existing molecular graphs are progressively corrupted.
2. The AI learns to predict how to restore the missing or altered information.
3. During generation, it starts from a noisy representation and repeatedly refines it.
4. It produces a candidate molecular graph.

Suppose the target connection is:-
Carbon → Nitrogen → Oxygen
But the starting graph contains the wrong connection:
Carbon → Oxygen → Nitrogen
A toy program might manually change the connections:
Step	What happens
1	Start with the incorrect graph.
2	Replace one connection with the target connection.
3	Replace another connection to complete the target graph.

For oncology research, a graph might represent a molecule:
- Nodes = atoms
- Edges = chemical bonds
- Node features = atom types
- Edge features = bond types

Python program = This example uses a tiny molecule graph. It demonstrates iterative refinement, not a trained diffusion model 
or a valid molecular generator.
"""
# Toy molecular graph for oncology drug-discovery research.
# Nodes represent atoms; edges represent bonds.
true_edges = {
    ("C1", "C2"),
    ("C2", "N1"),
    ("N1", "C3")
}
candidate_edges = {
    ("C1", "C2"),
    ("C1", "N1"),                   # Incorrect candidate connection
    ("N1", "C3")
}
def normalize_edge(edge):
    return tuple(sorted(edge))

true_edges = {normalize_edge(e) for e in true_edges}
candidate_edges = {normalize_edge(e) for e in candidate_edges}
# Simulated denoising steps.# In a real model, a trained neural network predicts # which edges should be added or removed.
steps = [{
        "remove"    : {("C1", "N1")},
        "add"       : {("C2", "N1")}
    },
    {
        "remove"    : set(),
        "add"       : set()
    }
]
print("Initial candidate:", sorted(candidate_edges))
for step_number, step in enumerate(steps, start=1):
    candidate_edges -= {normalize_edge(e) for e in step["remove"]}
    candidate_edges |= {normalize_edge(e) for e in step["add"]}
    print(f"After refinement {step_number}:")
    print(sorted(candidate_edges))
print("\nMatches example target:", candidate_edges == true_edges)
```
