```python
#---Causal Discovery from Graphs----
"""
Imagine the following possible relationships:
    Disease Severity → Treatment → Outcome
    Disease Severity → Outcome
If we observe that patients receiving a treatment have worse outcomes, we cannot immediately conclude that the treatment 
caused harm. More severely ill patients may have been more likely to receive that treatment.
Is a biomarker associated with cancer progression directly, or might the relationship be 
influenced by treatment and disease severity?

Causal discovery attempts to infer plausible relationships from observed data under specific assumptions.
Python program below-- The correlations help identify associations in the synthetic data. They do not discover the causal 
graph automatically.

For this example, we already know how the data was generated:
- Greater severity increases the probability of receiving treatment.
- Greater severity reduces the outcome score.
- Treatment increases the outcome score in this simulation.

The treatment-outcome correlation may therefore be misleading because treatment selection depends on severity.A more rigorous 
causal-discovery program could use methods such as PC or GES, with appropriate assumptions and statistical tests.
"""
import random
import math

random.seed(42)
patients = []
for _ in range(500):                                                # Generate illustrative synthetic patient records.
    severity = random.random()
    treatment_probability = 0.15 + 0.70 * severity                  # More severe patients are more likely to receive treatment.
    treatment = int(random.random() < treatment_probability)
    outcome_score = (                                               # Outcome depends on severity and treatment.
        0.8 - 0.7 * severity + 0.25 * treatment + random.gauss(0, 0.15))
    patients.append({
        "severity"  : severity,
        "treatment" : treatment,
        "outcome"   : outcome_score
    })
def correlation(records, x, y):
    xs = [row[x] for row in records]
    ys = [row[y] for row in records]
    mean_x = sum(xs) / len(xs)
    mean_y = sum(ys) / len(ys)
    numerator = sum((a - mean_x) * (b - mean_y) for a, b in zip(xs, ys))
    denominator = math.sqrt( sum((a - mean_x) ** 2 for a in xs) * sum((b - mean_y) ** 2 for b in ys))
    return numerator / denominator if denominator else 0
print(
    "Severity vs treatment:",round(correlation(patients, "severity", "treatment"), 3))
print(
    "Treatment vs outcome:", round(correlation(patients, "treatment", "outcome"), 3))
print(
    "Severity vs outcome:",round(correlation(patients, "severity", "outcome"), 3))
```
