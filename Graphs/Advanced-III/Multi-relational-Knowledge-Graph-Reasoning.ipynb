```python
#----Multi-relational Knowledge Graph Reasoning
# --How can we connect mutations, pathways, drugs, scientific publications, and observed outcomes using different 
# relationship types?
"""
A simple graph may represent every connection identically. A multi-relational knowledge graph explicitly distinguishes 
relationships such as:
- MUTATION_IN_GENE
- ACTIVATES_PATHWAY
- TARGETED_BY
- SUPPORTED_BY_PUBLICATION
- OBSERVED_WITH_OUTCOME
The type of relationship matters because each has a different meaning.

Oncology application: Connect molecular findings to pathways, drug targets, supporting publications, 
and separately documented clinical outcomes.

Python program -- The graph stores relationships as triples: source, relationship type, and target. The program can filter 
a particular type of relationship or search for a path to another entity.
"""
# Oncology research knowledge graph. Each edge is: (source, relationship, target).

triples = [
    ("EGFR_L858R", "VARIANT_OF", "EGFR"),
    ("EGFR", "PARTICIPATES_IN", "EGFR_Signaling"),
    ("Osimertinib", "TARGETS", "Mutant_EGFR"),
    ("Study_101", "REPORTS", "Osimertinib"),
    ("Study_101", "REPORTS", "Tumor_Response")
]

# Add an explicit link between the variant and its protein target.
triples.append(("EGFR_L858R", "REPRESENTED_BY", "Mutant_EGFR"))

def find_targets(start, relationship):
    return [target for source, rel, target in triples if source == start and rel == relationship]

def find_paths(start, target, max_depth=4):
    results = []
    def dfs(node, path, depth):
        if node == target:
            results.append(path)
            return
        if depth == max_depth:
            return
        for source, rel, next_node in triples:
            if source == node and next_node not in [step[2] for step in path]:
                dfs(next_node, path + [(source, rel, next_node)], depth + 1)
    dfs(start, [], 0)
    return results

print("Direct gene relationship:")
print(find_targets("EGFR_L858R", "VARIANT_OF"))
print("\nPaths from variant to protein target:")

for path in find_paths("EGFR_L858R", "Mutant_EGFR"):
    for source, rel, target in path:
        print(f"{source} --[{rel}]--> {target}")
```
