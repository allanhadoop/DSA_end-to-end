```python
#----Explainable GNNs and Subgraph Explanations---
"""
A Graph AI model makes a prediction, and Explainable GNN helps us understand which connected pieces of information 
influenced that prediction.

Oncology application: Show reviewers which source-linked findings and relationships influenced a research prediction, 
including missing evidence and conflicting records.

Real techniques include:
- GNNExplainer: Searches for a compact subgraph and feature subset that preserves a model's prediction.
- Integrated Gradients: Estimates how input features influence a model's output.
- Counterfactual explanations: Explore how a prediction changes when selected inputs or graph relationships are modified.

## What is subgraph explanation means -- 
A patient-history graph could contain hundreds of findings, tests, treatments, and outcomes.
Instead of displaying everything, the system highlights just the relevant connected pieces.

Full patient graph
Many records and relationships
Diagnosis---EGFR test--CT scan--Treatment---Blood test---Response---Comorbidity---Follow-up

Highlighted subgraph explanation
Only the findings relevant to the particular prediction ---
        EGFR finding
        Targeted therapy
        Documented response
This smaller section is called a subgraph: a selection of nodes and connections from a larger graph.

------
Python program-What does this demonstrate?
The program ranks the evidence items by their assigned contributions. It illustrates how an explanation could highlight 
a small subgraph rather than showing the entire patient-history graph.
But this is not a genuine explainable GNN: the contributions are manually assigned, not calculated by a trained model.

"""
# Example oncology research graph.# Contribution scores are illustrative and manually assigned.
evidence = {
    "EGFR_Mutation"             : 0.30,
    "EGFR_Mutation_to_Therapy"  : 0.25,
    "Treatment_Record"          : 0.20,
    "Documented_Response"       : 0.15,
    "Missing_Toxicity_Record"   : -0.05
}

baseline = 0.20                                             # Illustrative baseline score.
prediction_score = baseline + sum(evidence.values())
print(f"Illustrative model score: {prediction_score:.2f}")
print("\nEvidence ranked by contribution:")
for item, contribution in sorted(
    evidence.items(),
    key=lambda entry: abs(entry[1]),
    reverse=True
):
    print(f"{item}: contribution {contribution:+.2f}")
print("\nTop two influential evidence items:")
top_two = sorted(
    evidence.items(),
    key=lambda entry: abs(entry[1]),
    reverse=True
)[:2]
for item, contribution in top_two:
    print(f"- {item} ({contribution:+.2f})")
```
