```python
#-----Out-of-Distribution (OOD) Graph Generalization--A model was developed using data from Hospital A. How do we detect when Hospital B's data looks different?
"""
Hospitals can differ in patient populations, testing methods, referral patterns, coding practices, and treatment protocols.

A model may perform well at one hospital but poorly at another. An OOD detector attempts to identify inputs that differ 
from the data the model was trained on.

The biomarker-connection feature differs substantially between the two hospitals, so it should be flagged for investigation. 
The treatment-connection feature is much more similar. This is only a basic screening heuristic. It compares feature means 
and scales the difference by the training-set standard deviation; it does not establish that the new hospital is 
truly outside the training distribution.
"""
# Illustrative graph-derived features from two hospitals. # These values are invented for demonstration, not real patient data.

training_hospital = {
    "biomarker_connections": [2, 3, 2, 4, 3, 2],
    "treatment_connections": [1, 2, 1, 2, 1, 2]
}
new_hospital = {
    "biomarker_connections": [8, 9, 7, 10, 8, 9],
    "treatment_connections": [1, 2, 2, 1, 2, 1]
}
def mean(values):
    return sum(values) / len(values)
def population_std(values):
    avg = mean(values)
    return (sum((x - avg) ** 2 for x in values) / len(values)) ** 0.5
def compare_feature(reference, new_values):
    reference_mean = mean(reference)
    reference_std = population_std(reference)
    if reference_std == 0:
        if mean(new_values) == reference_mean:
            return 0.0
        return float("inf")
    return abs(mean(new_values) - reference_mean) / reference_std
for feature in training_hospital:
    distance = compare_feature(training_hospital[feature],new_hospital[feature])
    
    status = "FLAG FOR REVIEW" if distance > 2 else "Within screening range"
    print(feature)
    print(f"Standardized mean difference: {distance:.2f}")
    print(status)
    print()
```
