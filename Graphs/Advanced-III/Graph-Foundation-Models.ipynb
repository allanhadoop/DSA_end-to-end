```python
#----Graph Foundation Models----Can one pretrained model learn useful representations from many different kinds of graphs?
"""
A graph foundation model aims to learn reusable patterns from large collections of graph data, then adapt those 
learned representations to different tasks.
For oncology, those graphs might represent:
    - Molecular structures: atoms connected by chemical bonds.
    - Biomedical knowledge: genes connected to proteins or diseases.
    - Clinical relationships: documented biomarkers, treatments, and outcomes.
These are different graph types, so a single model does not automatically understand all of them.

--
Python program: A simplified transfer-learning demonstration
We'll create a tiny shared representation of biomedical concepts, then reuse it for two tasks: identifying molecular 
concepts and identifying clinical concepts. For example, a biomedical graph model could learn representations from many 
molecular relationships and then adapt them for a task such as identifying clinical relationships in oncology records.

So, the current code demonstrates the idea of reusing representations, but it does not demonstrate genuine pretrained-model 
transfer learning. Imagine a pretrained AI model has learned information about biomedical concepts. A graph can represent relationships explicitly
The pretrained model provides learned patterns from its training data. The graph provides an explicit structure for 
connecting entities and relationships.
    1) Creates a small graph manually
    2) Counts graph connections
    3) Ranks concepts by connection count

Flow---
    Step 1: Build a biomedical graph
    Connect EGFR, Osimertinib, Lung Cancer and Tumor Response.

    Step 2: Create shared representations
    Count how many connections each concept has.

    Step 3: Reuse those representations
    Apply the same connection counts to two different tasks.

        Task A
        Rank molecular concepts

        Task B
        Rank clinical concepts

"""
# Small illustrative biomedical graph.
graph = {
    "EGFR"              : ["Osimertinib", "Lung_Cancer"],
    "Osimertinib"       : ["EGFR"],
    "Lung_Cancer"       : ["EGFR", "Tumor_Response"],
    "Tumor_Response"    : ["Lung_Cancer"],
    "Protein_A"         : ["Protein_B"],
    "Protein_B"         : ["Protein_A"]
}
# Toy reusable representation: # number of connections for each concept. # A real foundation model would learn richer embeddings.
embeddings = { node : len(neighbors) for node, neighbors in graph.items()}

# Two downstream tasks reuse the same representations.
def rank_by_connectivity(nodes):
    return sorted(nodes, key=lambda node: embeddings[node], reverse=True)

molecular_task = ["EGFR", "Protein_A", "Protein_B"]
clinical_task = ["Lung_Cancer","Osimertinib","Tumor_Response"]
print("Shared graph representations:")
print(embeddings)
print("\nMolecular task ranking:")
print(rank_by_connectivity(molecular_task))
print("\nClinical task ranking:")
print(rank_by_connectivity(clinical_task))
```
