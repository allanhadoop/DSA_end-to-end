```python
#-----Causal Graph Neural Networks---A standard GNN learns from relationships among connected nodes. A causal GNN attempts to 
# incorporate causal structure or objectives into that learning process.
"""
Can a graph-based model learn treatment-outcome relationships and then be evaluated at a different hospital?

For example:
    Disease Severity → Treatment → Outcome
    Biomarker → Treatment
    Biomarker → Outcome

The challenge is that hospitals differ in patient populations, treatment practices, and data quality. 
A model that works at one hospital may not generalize to another.

Python program: A simplified causal-graph-inspired model
This demonstration estimates an outcome using known synthetic causal factors. It compares performance across two hospitals. 
It is not a trained GNN; it illustrates why adjusting for disease severity matters before evaluating generalization.

How to understand it
1. The hospitals have different severity distributions.
2. Treatment selection depends on severity and biomarker status.
3. The illustrative predictor adjusts for those observed factors.
4. We measure its prediction error separately at each hospital.
The results show how evaluation can vary between populations, even when the same relationship is used to generate the data.
"""
import random
import math

random.seed(7)
def generate_hospital(n, hospital_name, severity_shift):
    records = []
    for _ in range(n):
        severity = min(1.0, max(0.0, random.random() + severity_shift))
        biomarker = int(random.random() < 0.45)
        treatment = int(                                    # Treatment selection varies with severity and biomarker.
            random.random() < min(0.95, 0.15 + 0.55 * severity + 0.20 * biomarker))
        outcome = (0.9 - 0.6 * severity + 0.2 * biomarker + 0.3 * treatment + random.gauss(0, 0.12))
        records.append({
            "hospital"  : hospital_name,
            "severity"  : severity,
            "biomarker" : biomarker,
            "treatment" : treatment,
            "outcome"   : outcome
        })
    return records
hospital_a = generate_hospital(500, "Hospital_A", 0.0)
hospital_b = generate_hospital(300, "Hospital_B", 0.2)

# Illustrative model: a simple linear predictor. # Coefficients are specified for teaching, not learned from data.
def predict_outcome(patient):
    return (0.9 - 0.6 * patient["severity"] + 0.2 * patient["biomarker"] + 0.3 * patient["treatment"])
def mean_absolute_error(records):
    errors = [ abs(row["outcome"] - predict_outcome(row)) for row in records]
    return sum(errors) / len(errors)
print( "Hospital A prediction error:", round(mean_absolute_error(hospital_a), 3))
print( "Hospital B prediction error:", round(mean_absolute_error(hospital_b), 3))
```
