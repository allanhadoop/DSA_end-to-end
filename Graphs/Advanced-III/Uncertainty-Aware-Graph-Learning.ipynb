```python
#-----Uncertainty-Aware Graph Learning---What if an AI model produces a prediction but has very little evidence to support it?
# Suppose a graph model estimates whether a documented treatment outcome is likely to be positive. If several independently
# trained models disagree, the system could flag the prediction for review.
"""
Python Program--- 
- Patient A: Models largely agree on a higher probability.
- Patient B: Models disagree substantially, so the system flags the case.
- Patient C: Models agree on a probability near 0.50, which is not a confident positive or negative result.
Notice an important distinction: low disagreement does not necessarily mean a prediction is correct. All models can 
agree and still be wrong. This example measures ensemble disagreement, one source of uncertainty. It does not measure
every kind of uncertainty, and its thresholds are illustrative rather than validated.
"""

# Illustrative model outputs for an oncology research task. # Each value represents one model's estimated probability
# of a documented positive outcome.

predictions = {
    "Patient_A": [0.82, 0.79, 0.85, 0.81, 0.83],
    "Patient_B": [0.20, 0.90, 0.35, 0.85, 0.45],
    "Patient_C": [0.48, 0.52, 0.50, 0.49, 0.51]
}
def analyze_uncertainty(probabilities):
    mean_prediction = sum(probabilities) / len(probabilities)
    variance = sum((p - mean_prediction) ** 2 for p in probabilities) / len(probabilities)
    
    uncertainty = variance ** 0.5                   #---- Population standard deviation of model predictions.
    if uncertainty > 0.20:
        status = "HIGH disagreement: review evidence"
    elif uncertainty > 0.05:
        status = "MODERATE disagreement: inspect evidence"
    else:
        status = "LOW disagreement: still validate"
    return mean_prediction, uncertainty, status

for patient, probabilities in predictions.items():
    mean, uncertainty, status = analyze_uncertainty(probabilities)

    print(f"\n{patient}")
    print(f"Mean prediction: {mean:.2f}")
    print(f"Model disagreement: {uncertainty:.3f}")
    print("Status:", status)
```
