```python
#-----Matching algorithms -(Bipartite matching approach)--- Which patients can be matched to which clinical trials based on eligibility criteria?
# Imagine that two patients and two trials are available. A patient should only be matched to a trial if the patient 
# meets its defined eligibility criteria. This is a maximum bipartite matching approach. Each patient and each trial can appear in at most one match in this example.


"""
In practice, eligibility must consider inclusion and exclusion criteria, biomarkers, disease stage, prior therapies, geography, 
recruitment status, and trial capacity. Matching does not itself establish medical eligibility or consent.
----
What happened in the program?
1. Patient A was initially eligible for Trial 1.
2. Patient B also needed Trial 1.
3. The algorithm checked whether Patient A could be reassigned to Trial 2.
4. Since Patient A was eligible for Trial 2, both patients received distinct trial assignments.
"""
# Illustrative eligibility has already been checked.
# Each patient lists the trials for which they are eligible.

eligible_trials = {
    "Patient_A": ["Trial_1", "Trial_2"],
    "Patient_B": ["Trial_1"]
}
trial_to_patient = {}

def assign_trial(patient, visited_trials):
    for trial in eligible_trials[patient]:
        if trial in visited_trials:
            continue
        visited_trials.add(trial)            
        if (                                               # If the trial is free, or its current patient # can be moved to another eligible trial:
            trial not in trial_to_patient
            or assign_trial(
                trial_to_patient[trial],
                visited_trials
            )
        ):
            trial_to_patient[trial] = patient
            return True
    return False

for patient in eligible_trials:
    assign_trial(patient, set())
print("Patient-to-trial assignments:")

for trial, patient in trial_to_patient.items():
    print(patient, "->", trial)
print("Total matched patients:", len(trial_to_patient))
```
