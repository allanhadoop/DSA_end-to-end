```python
#----Eulerian and Hamiltonian paths
"""
Eulerian path = Visit every road once.
Hamiltonian path = Visit every location once.
"""
# Undirected road network connecting medical facilities.
roads = {
    "Warehouse"     : ["Clinic_A", "Clinic_B"],
    "Clinic_A"      : ["Warehouse", "Clinic_B"],
    "Clinic_B"      : ["Warehouse", "Clinic_A", "Hospital"],
    "Hospital"      : ["Clinic_B", "Clinic_C"],
    "Clinic_C"      : ["Hospital"]
}

def find_eulerian_path(graph):                                              # A. Eulerian path: use every road once
    degree = {node: len(neighbors) for node, neighbors in graph.items()}    # Creates a new dictionary using the node as the key and its connection count as the value.
    odd_nodes = [
        node for node, count in degree.items()
        if count % 2 == 1
    ]
    if len(odd_nodes) not in (0, 2):
        return None
    start = odd_nodes[0] if odd_nodes else next(iter(graph))
    edges = []                                                            # Each undirected road gets a unique ID.
    adjacency = {node: [] for node in graph}
    for u in graph:
        for v in graph[u]:
            if u < v:
                edge_id = len(edges)
                edges.append((u, v))
                adjacency[u].append((v, edge_id))
                adjacency[v].append((u, edge_id))
    used = set()
    stack = [start]
    path = []
    while stack:
        node = stack[-1]
        while adjacency[node] and adjacency[node][-1][1] in used:
            adjacency[node].pop()
        if adjacency[node]:
            neighbor, edge_id = adjacency[node].pop()
            used.add(edge_id)
            stack.append(neighbor)
        else:
            path.append(stack.pop())
    path.reverse()

    if len(used) != len(edges):
        return None
    return path

# B. Hamiltonian path: visit every facility once
def find_hamiltonian_path(graph):
    def search(node, path, visited):
        if len(path) == len(graph):
            return path
        for neighbor in graph[node]:
            if neighbor not in visited:
                result = search(
                    neighbor,
                    path + [neighbor],
                    visited | {neighbor}
                )
                if result is not None:
                    return result
        return None
    for start in graph:
        result = search(start, [start], {start})
        if result is not None:
            return result
    return None

euler_path = find_eulerian_path(roads)
hamilton_path = find_hamiltonian_path(roads)

print("Eulerian path:", euler_path)
print("Hamiltonian path:", hamilton_path)
```
