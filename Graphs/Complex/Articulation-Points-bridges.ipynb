```python
#-----Articulation points and bridges -- A hospital network can identify where backup referral arrangements may be needed.
# Important: this program assumes undirected connections. Real referral networks may be directional, and 
# practical criticality also depends on capacity, travel time, and available alternatives.
"""
The program uses DFS (Depth-First Search) to explore connected hospitals. It checks whether another route exists around each node 
or connection.
- If removing Referral_Hub disconnects several hospitals, it is an articulation point.
- If removing the connection between Nashik_Hospital and Aurangabad_Hospital isolates Aurangabad, that connection is a bridge.

Healthcare question: Which hospital or referral connection is critical to keeping oncology care connected?
- Articulation point: A hospital or facility whose removal disconnects parts of the network.
- Bridge: A connection whose removal disconnects parts of the network.
"""
graph = {
    "Pune_Hospital"         : ["Satara_Hospital"],
    "Satara_Hospital"       : ["Pune_Hospital","Referral_Hub"],
    "Referral_Hub"          : ["Satara_Hospital","Mumbai_Cancer_Center","Nashik_Hospital"],
    "Mumbai_Cancer_Center"  : ["Referral_Hub"],
    "Nashik_Hospital"       : ["Referral_Hub", "Aurangabad_Hospital"],
    "Aurangabad_Hospital"   : ["Nashik_Hospital"]
}

def find_critical_points_and_bridges(graph):
    discovery = {}
    low = {}
    parent = {}
    articulation_points = set()
    bridges = []
    time = [0]

    def dfs(node):
        discovery[node] = low[node] = time[0]
        time[0] += 1
        children = 0
        for neighbor in graph[node]:
            if neighbor not in discovery:
                parent[neighbor] = node
                children += 1
                dfs(neighbor)

                low[node] = min(low[node], low[neighbor])
                if low[neighbor] > discovery[node]:             # Is this connection a bridge?
                    bridges.append((node, neighbor))

                if node not in parent:                          # Can removing this node disconnect the network?
                    if children > 1:
                        articulation_points.add(node)
                elif low[neighbor] >= discovery[node]:
                    articulation_points.add(node)

            elif neighbor != parent.get(node):
                low[node] = min(low[node], discovery[neighbor])

    for node in graph:
        if node not in discovery:
            dfs(node)
    return articulation_points, bridges

points, bridges = find_critical_points_and_bridges(graph)
print("Articulation points:", sorted(points))
print("Bridges:", sorted(tuple(sorted(edge)) for edge in bridges))  #- Converts the sorted list back into a tuple. Double sort
""" 
Code	                            Meaning
--------                            -----------
sorted(edge)	                    Sorts the two records inside one connection.
tuple(sorted(edge))	                Sorts the records, then converts them to a tuple.
sorted(... for edge in bridges)	    Processes every connection and sorts the final list.
"""
```
