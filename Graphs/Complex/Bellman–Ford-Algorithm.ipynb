```python
#----Bellman–Ford algorithm 
"""
Suppose a cancer patient needs to be transferred between healthcare facilities, and each connection has a cost.
Some connections may have negative weights because a healthcare network applies a credit, reimbursement, or cost adjustment.
Bellman–Ford finds the lowest total cost from one starting node to other reachable nodes, even when some edges have negative 
weights. It also detects reachable negative-weight cycles, where no finite minimum path exists.
---
Route	                Initial or direct cost	        Improved cost	        Reason
Pune → Diagnostic Lab	8	                            3	                    Pune → Regional Hub → Lab costs \(5 + (-2) = 3\)
Pune → Regional Hub	    5	                            5	                    Already the cheapest known route
Pune → Mumbai	        15 via Regional Hub	            7	                    Pune → Regional Hub → Lab → Mumbai costs \(5-2+4=7\)
"""
edges = [
    ("Pune_Cancer_Center", "Diagnostic_Lab", 8),
    ("Pune_Cancer_Center", "Regional_Oncology_Hub", 5),
    ("Regional_Oncology_Hub", "Diagnostic_Lab", -2),              # <<------Negative weight  (e.g credit. reimbursement etc)
    ("Diagnostic_Lab", "Mumbai_Hospital", 4),
    ("Regional_Oncology_Hub", "Mumbai_Hospital", 10)
]
nodes = [
    "Pune_Cancer_Center",
    "Diagnostic_Lab",
    "Regional_Oncology_Hub",
    "Mumbai_Hospital"
]
source = "Pune_Cancer_Center"
distance = {node: float("inf") for node in nodes}        # Step 1: Set all distances to infinity
distance[source] = 0                                     # Starting point has cost zero
for _ in range(len(nodes) - 1):                          # Step 2: Relax all edges repeatedly
    changed = False
    for u, v, weight in edges:
        if distance[u] != float("inf"):
            new_cost = distance[u] + weight
            if new_cost < distance[v]:
                distance[v] = new_cost
                changed = True
    if not changed:
        break

negative_cycle = False                          # Step 3: Check for a reachable negative cycle
for u, v, weight in edges:
    if (distance[u] != float("inf")
            and distance[u] + weight < distance[v]):
        negative_cycle = True
        break
if negative_cycle:
    print("Reachable negative-weight cycle detected")
else:
    print("Minimum costs from", source)
    for node, cost in distance.items():
        print(node, ":", cost)
```
