```python
#----Floyd–Warshall algorithm 
"""
Bellman–Ford calculates shortest paths from one starting node.
Floyd–Warshall calculates shortest paths between every pair of nodes.
For example, a healthcare network might want to know the cheapest route between Pune and Mumbai, Pune and Nashik, 
Mumbai and Nashik, and every other pair.
The same calculation works for every pair of hospitals, which is why the algorithm uses three loops:
- k: the possible intermediate hospital.
- i: the starting hospital.
- j: the destination hospital.

Bellman–Ford is useful when you need paths from one source; Floyd–Warshall is useful when you need paths between all pairs. 
Floyd–Warshall can handle negative edges, but a negative cycle prevents finite shortest-path results for affected pairs.
"""

INF = float("inf")
hospitals = ["Pune", "Nashik", "Mumbai"]
distance = [                           # Initial direct-route costs # INF means there is no direct connection
    [0,   8,  20],  # Pune
    [8,   0,   5],  # Nashik
    [20,  5,   0]   # Mumbai
]
n = len(hospitals)
for k in range(n):                      # Consider every hospital as an intermediate stop
    for i in range(n):
        for j in range(n):
            new_cost = distance[i][k] + distance[k][j]
            if new_cost < distance[i][j]:
                distance[i][j] = new_cost
print("Shortest route cost matrix:")    # Display the final shortest-distance matrix
print("       ", hospitals)

for i in range(n):
    print(hospitals[i], distance[i])    # Example: shortest cost from Pune to Mumbai
print("Pune to Mumbai:", distance[0][2])    
```
