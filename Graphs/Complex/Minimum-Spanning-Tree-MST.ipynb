```python
#----Minimum Spanning Tree (MST) -- Kruskal's algorithm
# Suppose five oncology facilities need to be connected to a common healthcare data-sharing network.
# You want every facility to be connected, but you want the total connection cost to be as low as possible.
# A Minimum Spanning Tree connects all nodes in a connected, undirected, weighted graph without creating cycles, 
# using the minimum total edge weight.
"""
The algorithm rejects a connection if its endpoints are already connected, 
because adding that edge would create a cycle. The total cost is \(2+3+4=9\).

The edge weights represent hypothetical infrastructure costs for secure data connectivity. Kruskal's algorithm is one way to 
find an MST. It sorts connections by cost and selects the cheapest connection that does not create a cycle.
"""
edges = [
    (2, "Pune", "Nashik"),                 # Cost frpm Pune to Nashik
    (4, "Nashik", "Mumbai"),
    (3, "Pune", "Satara"),
    (6, "Satara", "Mumbai"),
    (5, "Pune", "Mumbai")
]
parent = {                                 # Start with each hospital as its own group
    "Pune": "Pune",
    "Nashik": "Nashik",
    "Mumbai": "Mumbai",
    "Satara": "Satara"
}
def find(node):
    if parent[node] != node:
        parent[node] = find(parent[node])
    return parent[node]
def union(a, b):
    root_a = find(a)
    root_b = find(b)
    if root_a == root_b:
        return False                           
    parent[root_b] = root_a               # Would create a cycle
    return True

edges.sort()                              # Process lowest-cost connections first
mst = []
total_cost = 0

for cost, hospital_a, hospital_b in edges:
    if union(hospital_a, hospital_b):
        mst.append((hospital_a, hospital_b, cost))
        total_cost += cost
print("Selected connections:")
for a, b, cost in mst:
    print(a, "<->", b, ":", cost)
print("Total network cost:", total_cost)
```
