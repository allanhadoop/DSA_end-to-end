```python
# ---------Strongly Connected Components (SCC)
# A Strongly Connected Component is a group of nodes in a directed graph where every node can reach every other node 
# by following the arrow directions.
"""
A → B → C
↑       |
└───────┘
Every state must be able to reach every other state by following the arrows.
Starting state	        Can reach
--------------          --------
A	                    B and C
B	                    C and A
C	                    A and B
---------Example 2 - 
A → B → C
    A can reach B and C.
    B can reach C, but cannot reach A.
    C cannot reach A or B.
"""
graph = {
    "Review"        : ["Investigate"],
    "Investigate"   : ["Tumor_Board"],
    "Tumor_Board"   : ["Review", "Treatment"],
    "Treatment"     : ["Follow_Up"],
    "Follow_Up"     : ["Treatment"]
}

index = 0
stack = []
on_stack = set()
indices = {}
low = {}
components = []

def strongconnect(node):
    global index
    indices[node] = index                                   # ---  Assign discovery index and low-link value
    low[node] = index
    index += 1

    stack.append(node)
    on_stack.add(node)

    for neighbor in graph[node]:
        if neighbor not in indices:
            strongconnect(neighbor)                         # ---- Visit an unvisited neighbor
            low[node] = min(low[node], low[neighbor])
        elif neighbor in on_stack:
            low[node] = min(low[node], indices[neighbor])    #---- Back edge to a node still on the stack
    if low[node] == indices[node]:                           # If node is the root of an SCC, pop that component
        component = []
        while True:
            member = stack.pop()
            on_stack.remove(member)
            component.append(member)
            if member == node:
                break
        components.append(component)

for node in graph:                                           # Start DFS from each unvisited node
    if node not in indices:
        strongconnect(node)
print("Strongly Connected Components:")
for component in components:
    print(component)
```
