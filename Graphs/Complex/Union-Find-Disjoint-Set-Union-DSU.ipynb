```python
# --- Union-Find / Disjoint Set Union (DSU)
# Union-Find keeps track of which items belong to the same group.
# It supports two important operations:
# - Find: Identify which group an item belongs to.
# - Union: Merge two groups when a verified relationship shows they should belong together.
"""
A hospital may have several records associated with the same patient journey:
- Record_A — lung biopsy
- Record_B — molecular test
- Record_C — oncology consultation
- Record_D — another patient's record

Suppose verified identity matching establishes that A and B belong together, and B and C belong together. 
Union-Find can maintain this group without repeatedly searching every record.
Important: records should be merged only through an authorized, verified identity-matching process. Similar cancer 
diagnoses or mutations alone do not establish that two records belong to the same person.

The find() function identifies a representative for each group. The union() function combines groups when they have 
different representatives. Path compression in find() makes future lookups faster.
"""

parent = {
    "Record_A": "Record_A",
    "Record_B": "Record_B",
    "Record_C": "Record_C",
    "Record_D": "Record_D"
}
def find(record):
    if parent[record] != record:                         # Follow parent links to find the group representative
        parent[record] = find(parent[record])
    return parent[record]
def union(record1, record2):
    root1 = find(record1)
    root2 = find(record2)
    if root1 != root2:
        parent[root2] = root1
        print(record1, "and", record2, "grouped together")
    else:
        print("Already in the same group")

union("Record_A", "Record_B")                            # Verified identity matches
union("Record_B", "Record_C")

print("A group:", find("Record_A"))                      # Check which records belong to the same group
print("B group:", find("Record_B"))
print("C group:", find("Record_C"))
print("D group:", find("Record_D"))

print("A and C same group:",                            # Check whether A and C are in the same group
      find("Record_A") == find("Record_C"))
```
