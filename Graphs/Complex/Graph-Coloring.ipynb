```python
#-----Graph coloring is a way to schedule activities so that activities that conflict do not happen at the same time.
# 1. Imagine three activities
"""
        - Chemotherapy_A
        - Chemotherapy_B
        - CT_Scan
The hospital has two scheduling problems:
- Chemotherapy A and B use the same chair, so they cannot happen at the same time.
- Chemotherapy A and the CT scan need the same specialist, so they also cannot happen at the same time.

In graph coloring:
- Each node represents a procedure or appointment.
- An edge connects two procedures that cannot occur in the same time slot.
- A color represents a time slot.
- Connected nodes must receive different colors.

Chemotherapy A cannot share its time slot with either other activity, but Chemotherapy B and the CT scan can overlap 
under the simplified assumptions.
The program uses a greedy algorithm: it assigns each activity the first available color. It produces a valid schedule 
for this conflict graph, but it does not always find the minimum possible number of time slots.
"""
# An edge means two activities cannot share a time slot.
conflicts = {
    "Chemotherapy_A"    : ["Chemotherapy_B", "CT_Scan"],
    "Chemotherapy_B"    : ["Chemotherapy_A"],
    "CT_Scan"           : ["Chemotherapy_A"]
}

def color_graph(graph):
    assigned = {}
    for activity in graph:
        used_colors = set()
        for neighbor in graph[activity]:              # ---Collect time slots already used by conflicting activities.
            if neighbor in assigned:
                used_colors.add(assigned[neighbor])
        color = 0                                
        while color in used_colors:                   # --- Choose the first available time slot.
            color += 1
        assigned[activity] = color
    return assigned
schedule = color_graph(conflicts)

for activity, slot in schedule.items():
    print(f"{activity}: Time slot {slot + 1}")
```
