```python
#-----Probabilistic Graphical Models--- If a biomarker result is missing, how can we represent uncertainty rather than treating the answer as definitely positive or negative?
"""
Suppose a clinical record does not contain a usable EGFR result. A probabilistic model can represent uncertainty about the result using prior information and relevant evidence.
For a simple illustration, assume:
- Prior probability of a particular biomarker state: 30%
- Probability of a positive test if that state is present: 90%
- Probability of a positive test if that state is absent: 10%
We can calculate the probability of the state after observing a positive test using Bayes' theorem.

Before the test, the assumed probability was 30%. Given the assumed test characteristics, a positive result raises the 
calculated probability to approximately 79.4%.

"""
# Illustrative probabilities, not real clinical estimates.
prior_positive = 0.30
sensitivity = 0.90                                   #----- Test characteristics assumed for this example.
false_positive_rate = 0.10

p_test_positive_if_state = sensitivity               #----- Probability of a positive test under each state.
p_test_positive_if_not_state = false_positive_rate

p_test_positive = (                                  # ---- Total probability of a positive test.
    prior_positive * p_test_positive_if_state
    + (1 - prior_positive) * p_test_positive_if_not_state
)

# Bayes' theorem.
posterior_positive = ( prior_positive * p_test_positive_if_state / p_test_positive)

print("Prior probability:", f"{prior_positive:.1%}")
print("Probability after a positive test:",
      f"{posterior_positive:.1%}")
```
