```python
#--------Hypergraphs--------In a normal graph, an edge usually connects two nodes. In a hypergraph, one hyperedge can connect three or more nodes together.
# How can we represent one clinical relationship that involves several entities at the same time?
"""
Imagine a documented oncology case involving:
- A patient with an EGFR mutation
- A treatment with osimertinib
- A clinical context describing the disease and treatment line
- A recorded tumor response
These entities belong to one clinical episode. Representing the episode as one hyperedge preserves their joint relationship.

Episode_101 is returned by below program because it contains both the EGFR mutation and osimertinib, together with the other 
entities and its source.

"""
# A simple oncology hypergraph. # Each hyperedge connects multiple entities simultaneously.
hyperedges = [
    {
        "episode_id": "Episode_101",
        "entities": {
            "Patient_101",
            "EGFR_L858R",
            "Osimertinib",
            "Advanced_Lung_Cancer",
            "Recorded_Response"
        },
        "source": "Oncology_Note_101"
    },
    {
        "episode_id": "Episode_102",
        "entities": {
            "Patient_102",
            "HER2_Positive",
            "Targeted_Therapy",
            "Breast_Cancer",
            "Recorded_Progression"
        },
        "source": "Oncology_Note_102"
    }
]
def find_episodes_containing(required_entities):
    matches = []
    for edge in hyperedges:
        if required_entities.issubset(edge["entities"]):                # .issubset(...) — checks whether every required entity exists in that episode.
            matches.append(edge)
    return matches

results = find_episodes_containing({ "EGFR_L858R","Osimertinib"})     # Find episodes containing both a mutation and a treatment.
for episode in results:
    print("Episode:", episode["episode_id"])
    print("Entities:", sorted(episode["entities"]))
    print("Evidence source:", episode["source"])
```
