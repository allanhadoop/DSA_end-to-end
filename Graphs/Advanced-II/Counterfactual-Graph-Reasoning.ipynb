```python
#---------Counterfactual Graph Reasoning---For the same patient, how might the outcome have differed if a different treatment had been given?
"""
This is different from asking, “What happened to patients who received Drug A versus Drug B?”

    A counterfactual asks:
    For a particular patient who received Drug A, what might have happened to that same patient if they had instead received Drug B?

We cannot observe both outcomes for the same patient at the same time. We therefore need a causal model and assumptions to 
estimate the alternative scenario. This is a model-generated hypothetical, not evidence that Drug B would actually improve this patient's outcome by five points.
"""

# Illustrative structural causal model. # This is NOT a clinically valid treatment-effect model.

def predict_outcome(severity, treatment):
    """
    severity: 0 = lower severity, 1 = higher severity
    treatment: 'A' or 'B'
    Returns a synthetic outcome score from 0 to 100.
    """
    baseline = 80 - (30 * severity)
    treatment_effect = {"A": 5,"B": 10}
    score = baseline + treatment_effect[treatment]
    return max(0, min(100, score))

#------ Consider the same hypothetical patient with high severity.
severity = 1
observed_treatment = "A"
alternative_treatment = "B"
observed_scenario = predict_outcome(severity, observed_treatment)

alternative_scenario = predict_outcome(severity, alternative_treatment)
difference = alternative_scenario - observed_scenario

print("Observed treatment:", observed_treatment)
print("Model score under observed treatment:", observed_scenario)
print("\nAlternative treatment:", alternative_treatment)
print("Model score under alternative treatment:",
      alternative_scenario)
print("\nCounterfactual difference:", difference)
```
