```python
#-----Heterogeneous Graph Transformers (HGT)---Can AI learn relationships among different types of oncology information at the same time?
"""
A normal graph may contain only one type of node, such as hospitals. An oncology knowledge graph can contain many types:
- Patient
- Mutation
- Pathology finding
- Drug
- Clinical study
- Treatment outcome
The relationships are also different: a patient has a mutation, a drug targets a mutation, and a patient experienced an outcome.
---
        The graph distinguishes a patient from a mutation, drug, or outcome.
        Every relationship has a type, such as TARGETS.
        The program uses a different illustrative weight for each relationship type.
        Nodes combine their original values with information received from connected nodes.

What makes a real HGT different? A Heterogeneous Graph Transformer uses learned representations and attention mechanisms 
that can distinguish node types and relation types. The model learns how much attention to give different neighbors 
during training. Our program uses manually assigned weights, so it is not an actual transformer.
"""
# Simplified oncology heterogeneous graph.# This is NOT a trained HGT model.
nodes = {
    "Patient_1" : {"type": "patient", "value": 1.0},   #---Nested dictionary
    "EGFR"      : {"type": "mutation", "value": 0.9},
    "Drug_A"    : {"type": "drug", "value": 0.8},
    "Outcome_1" : {"type": "outcome", "value": 1.0}
}
edges = [
    ("Patient_1", "EGFR", "HAS_MUTATION"),
    ("Drug_A", "EGFR", "TARGETS"),
    ("Patient_1", "Outcome_1", "HAS_OUTCOME")
]
relation_weights = {                                        # Different relationship types get different illustrative weights.
    "HAS_MUTATION"  : 0.9,
    "TARGETS"       : 0.7,
    "HAS_OUTCOME"   : 0.8
}
messages = {node: [] for node in nodes}
for source, target, relation in edges:
    weight = relation_weights[relation]
    messages[target].append(nodes[source]["value"] * weight)  # Send information in both directions for this toy example.
    messages[source].append(nodes[target]["value"] * weight)
print("Node information after collecting messages:")
for node, data in nodes.items():
    incoming = messages[node]
    if incoming:
        neighbor_average = sum(incoming) / len(incoming)
        updated = (data["value"] + neighbor_average) / 2
    else:
        updated = data["value"]
    print(
        node,
        "| type:", data["type"],
        "| updated value:", round(updated, 3)
    )
```
