```python
#----Causal Graphs and Structural Causal Models (SCMs)
# If patients receiving a particular treatment have worse outcomes, does that mean the treatment caused the worse outcomes?
# Not necessarily. Patients with more severe disease may be more likely to receive that treatment and more likely 
# to have a poor outcome.

# Disease severity --- Influences---------Treatment received
# Disease severity --- Influences---------Outcome

#--- Disease severity may influence both treatment selection and the outcome.
"""
The overall figures make Drug A appear much worse. But Drug A was given to many more patients with severe disease.
Within each severity group, the favorable outcome rates are identical in this synthetic example.
The apparent overall difference is explained by the different mix of disease severity in the two treatment groups. This is 
an example of confounding, sometimes producing Simpson's paradox.

A Structural Causal Model goes further by specifying equations for how variables generate one another, including assumptions 
about treatment selection and outcomes. A causal graph represents those assumptions, and causal inference methods use 
suitable data and assumptions to estimate effects.
"""
# Synthetic data: 100 patients. # Outcome = 1 means a favorable outcome.  These numbers are invented for teaching.

patients = []
patients += [{"severity": "Mild", "drug": "A", "good": 8}]    # Mild disease: Drug A has 8 favorable outcomes out of 10.
patients += [{"severity": "Mild", "drug": "B", "good": 72}]
patients += [{"severity": "Severe", "drug": "A", "good": 18}] # Severe disease: Drug A has 18 favorable outcomes out of 90.
patients += [{"severity": "Severe", "drug": "B", "good": 2}]

groups = {                                                    # Store group counts as: # (number of favorable outcomes, total patients)
    ("Mild", "A")   : (8, 10),                                # means 8 favorable outcomes among 10 mild-disease patients who received Drug A.
    ("Mild", "B")   : (72, 90),                               # counts[0] = 72  and counts[1] = 90
    ("Severe", "A") : (18, 90),
    ("Severe", "B") : (2, 10)
}
for drug in ["A", "B"]:                                         # Crude comparison, ignoring severity.
    good = sum(counts[0] for (severity, d), counts in groups.items() if d == drug)          # Key: Disease severity and drug. and value is (favorable outcomes, total patients).
    total = sum(counts[1] for (severity, d), counts in groups.items()if d == drug)          # Find all entries for the selected drug and add their favorable outcome counts.
    print(f"Drug {drug}: {good}/{total} favorable " f"({good / total:.1%})")

print("\nCompare within severity groups:")
for severity in ["Mild", "Severe"]:
    for drug in ["A", "B"]:
        good, total = groups[(severity, drug)]
        print(severity, "| Drug", drug, "| favorable outcome:", f"{good / total:.1%}"
        )

"""
good = sum(
    counts[0]
    for (severity, d), counts in groups.items()
    if d == drug
)

This above sum can be written as below -
good = 0
for (severity, d), counts in groups.items():
    if d == drug:
        good = good + counts[0]
"""
```
