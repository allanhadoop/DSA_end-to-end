```python
#----Temporal Graph Neural Networks (TGNNs)-- Can AI learn not just which clinical events are connected, but how those connections change over time?
"""
Consider a simplified patient history:
- January: EGFR mutation documented
- February: targeted treatment started
- April: response assessed
- August: progression documented
A static graph records the connections. A temporal graph also records when events and relationships occurred.

The program processes events in time order and maintains a simple state: the currently recorded treatment.
When progression is recorded in August, the history shows that Drug A had previously been started. But this program is 
not yet a TGNN. It is a simple temporal data-processing example.

A genuine Temporal Graph Neural Network can learn from time-stamped interactions and changing node states. It might be 
trained to estimate a future event, such as the likelihood of documented progression within a specified time window, 
provided that the data and validation support that task.

"""
# Synthetic oncology event stream. # Dates and events are illustrative.
events = [
    ("2026-01-10", "EGFR_Mutation", "Documented"),
    ("2026-02-01", "Drug_A", "Treatment_Started"),            # <<<- new event change/added
    ("2026-04-15", "Tumor", "Response_Assessed"),
    ("2026-08-20", "Tumor", "Progression_Documented")
]
events.sort(key=lambda event: event[0])
patient_history = []
active_treatment = None
for date, subject, event_type in events:
    if event_type == "Treatment_Started":
        active_treatment = subject
    patient_history.append({
        "date": date,
        "subject": subject,
        "event": event_type,
        "active_treatment": active_treatment
    })
print("Chronological oncology history:\n")
for event in patient_history:
    print(
        event["date"],
        "|", event["event"],
        "|", event["subject"],
        "| active treatment:", event["active_treatment"]
    )
```
