```python
#-----Knowledge graph completion --If a relationship is missing from our oncology knowledge graph, can we identify a 
# possible link that should be investigated?
"""
A knowledge graph might contain:
- EGFR L858R is associated with lung cancer.
- Osimertinib targets EGFR.
- A patient's record documents EGFR L858R.
A system might identify the missing candidate relationship between that molecular finding and the documented treatment record.
Crucial distinction: A missing relationship is not necessarily a true relationship. Knowledge graph completion generates 
candidates for validation; it must not invent facts in a patient's clinical history.

Outcome of below program -- The program finds a plausible candidate because the patient has a recorded EGFR mutation and 
osimertinib targets EGFR. But the program has not established that Patient 101 actually received osimertinib. 
The missing edge could be true, false, or simply undocumented in the available graph.

"""
# Illustrative biomedical knowledge graph. # Relationships are represented as (subject, relation, object).

triples = {
    ("Patient_101", "has_mutation", "EGFR_L858R"),
    ("Osimertinib", "targets", "EGFR"),
    ("Patient_101", "has_diagnosis", "Lung_Cancer")
}
def has_triple(subject, relation, obj):
    return (subject, relation, obj) in triples
candidates = []                                             # Candidate relationships generated from a simple rule.
for patient, relation, mutation in triples:
    if relation != "has_mutation":
        continue
    gene = mutation.split("_")[0]                           # Look for drugs that target the same gene.
    for drug, drug_relation, target in triples:
        if drug_relation != "targets":
            continue
        if target == gene:
            candidate = (
                patient,
                "documented_treatment",
                drug
            )
            if not has_triple(*candidate):
                candidates.append(candidate)
print("Candidate missing relationships:")
for candidate in candidates:
    print(candidate)
```
