```python
#---------Create and connect Nodes
# Goal: Create three nodes and print their values.
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None

a = Node(10)
b = Node(20)
c = Node(30)

a.next = b         # << -- next is std. attribute name 
b.next = c

current = a        # << -- current is variable name of class object a 

while current is not None:
    print(current.data)
    current = current.next

#   a             b             c
# [10 | next] → [20 | next] → [30 | None]
```

```python
#---- Insert a node into a Linked List

class Node:
    def __init__(self, data):
        self.data = data
        self.next = None

a = Node(10)
b = Node(20)
c = Node(30)
a.next = b
b.next = c

new_node = Node(15)

new_node.next = a.next        # <<-------------Insert
a.next = new_node

current = a
while current is not None:
    print(current.data)
    current = current.next
```

```python
#---------Reverse a Linked List ---  Reverse the direction of all connections.

class Node:
    def __init__(self, data):
        self.data = data
        self.next = None

a = Node(10)
b = Node(20)
c = Node(30)

a.next = b
b.next = c
previous = None
current = a

while current is not None:
    next_node = current.next
    current.next = previous
    previous = current
    current = next_node

current = previous                  # previous is now the new head

while current is not None:
    print(current.data)
    current = current.next
```
