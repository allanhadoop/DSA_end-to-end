```python
# ----------Find duplicates and remove it -----------------
numbers = [10, 20, 10, 30, 20, 400, 10, 300]

singlentry = []
duplicate = []
count_of_duplicate = {}

for n in numbers:
    if not n in singlentry:
        singlentry.append(n)
    else:
        duplicate.append(n)
        if n in count_of_duplicate:
            count_of_duplicate[n] += 1
        else:
            count_of_duplicate[n] = 1

print(duplicate)
print(count_of_duplicate)

# ---------Remove duplicates -------------Problem: We have repeated values and want only unique values.
print("----------------Set----------------")
unique = set(numbers)                    # << -- This set will identify only unique values
print(unique)

#-------using hash map to remove duplicates 
hash_map_dic = {}
for n in numbers:
    hash_map_dic[n] = True               # << -- This will update hash_map_dic{} only with singles , no duplicates

print("----------------Hash Map----------------")
print(list(hash_map_dic.keys()))         # <<-- This is how we access keys 
```

```python
#--------------Group data -(List (Tuple)) ---------Group students by city 
students = [
    ("Rahul", "Pune"),            # <--- This is tuple so we donot use .items to access it.
    ("Priya", "Mumbai"),
    ("Amit", "Pune"),
    ("Neha", "Mumbai"),
    ("John", "Delhi")
]

groups = {}
for name, city in students:           # <--- Student has values in tuple so we donot use .items to access it.
    if city not in groups:
        groups[city] = []
    groups[city].append(name)
print(groups)
```

```python
#---- Group patients by disease ---(List--(Dic))---------

patients = [
    {'sam' : 'diabetic'},
    {'tim' : 'heart failure'},
    {'kim' : 'diabetic'},
    {'jim' : 'liver failed'},
    {'bim' : 'heart failure'}
]
group = {}
for patient in patients:                              # <<<-- For dic, use nested loop
    for name, disease in patient.items():             # <<<-- For dic, use .items() to access 
        if not disease in group: 
            group[disease] = []
        group[disease].append(name)

print(group)
```

```python
#---------Two sum problem----------(Hash Map problems in DSA)
#----The Hash Map acts like our memory of previously seen numbers.
numbers = [2, 10, 11, 7]
target = 9

seen = {}
for number in numbers:
    required = target - number
    if required in seen:
        print(required, "+", number, "=", target)
        break
    seen[number] = True          # <<--- The Hash Map acts like our memory of previously seen numbers.
```

```python
#-----------Cache previous results------
cache = {}
number = 10

if number in cache:
    result = cache[number]
else:
    result = number * number
    cache[number] = result

print(result)
```

```python
#----------Anagram Detection---Check whether two words contain exactly the same letters. hash map to count the letters: Hash Map role: letter → number of occurrences
# listen
# silent

word = "listen"
count = {}
for letter in word:
    count[letter] = count.get(letter, 0) + 1

print(count)
# listen → same letter counts → silent
# "listen" and "silent" → Anagrams ✅
```

```python
#-----------Sliding Window - Look at a small portion of data and move that portion forward.
# Remove the item leaving the window + add the item entering the window.
"""
First window:
100  200  150
└────────────┘
Move one position:
     200  150  300
     └────────────┘
Move again:
          150  300  250
          └────────────┘
"""
transactions = [
    "food",
    "travel",
    "food",
    "shopping",
    "food"
]
window_size = 3
window = {}
""" 
range(stop)              → start automatically = 0
range(start, stop)       → you specify start and stop
range(start, stop, step) → you specify start, stop and movement
"""
for item in transactions[:window_size]:               #----# Create first window
    window[item] = window.get(item, 0) + 1            #--- Get() is purposly used to count each word
print("Window 1:", window)

for i in range(window_size, len(transactions)):       #----# Slide the window
   
    leaving = transactions[i - window_size]           #----- # Remove item leaving the window
    window[leaving] -= 1

    entering = transactions[i]            
    window[entering] = window.get(entering, 0) + 1    #---- Add item entering the window

    print("Window :", window)
```

```python
## Basically we are playing with looping text of railway stations, then converted list into dic, then played with index to add or remove so looping on indexes
railwaystations = ['virar', 'nalasopara', 'vasai','naigaon','bhyandar','virar','dahisar','borivali']

size = 2
station = {}

# print till size 2
for i in railwaystations[:size]:
    station[i] = station.get(i,0) + 1
print("STation reached : ", station)

# print from 2 till end , while printing , take station count and remove if duplicate
#--- Here we are dealing more with add/remove station based on key or a number in station dic so we use range loop
#----and not railwaystation look which is text(station name)
for j in range(size ,len(railwaystations)):
    dropindex = railwaystations[j - size]
    station[dropindex] -= 1

    addstation = railwaystations[j]
    station[addstation] = station.get(addstation,0) + 1 

    print("Station reached : ", station)
```

```python
#---------Add 10 mins increment to time whose name is Borivali and if duplicate station then remove lesser stop time entry
var = [
    {'virar'     : 2},
    {'nalasopara': 2},
    {'vasai'     : 3},
    {'naigaon'   : 2.5},
    {'vasai'     : 5},
    {'Miraroad'  : 3},
    {'Borivali'  : 5}]

checkname = 'Borivali'
duplicate = {}

for i in var:
    for name, time in i.items():
        if name == checkname: 
            time = time + 10
        if not name in duplicate:
            print("----Qualified station:", name, "Stop time:", time)  
            duplicate[name] = time   
        else:
            if duplicate[name] <= time:
                print(f"\033[94m------Duplicate removed with less stop time \033[0m {name}-----", duplicate[name])
                duplicate.pop(name)
                duplicate[name] = time
```

```python
#-----------Advanced — LRU(Least Recently Used) Cache--------Keep only the data that has been used recently.---
#---LRU Cache = Keep recently used data and remove the data that hasn't been used for the longest time.
"""
Before:------
Google   YouTube   Amazon
After opening Netflix:-------
YouTube   Amazon   Netflix
   ↑
Google removed
----------
             HASH MAP
                 ↓
        "Remember something"
                 ↓
      ┌──────────┼──────────┐
      ↓          ↓          ↓
   Results     Users      Counts
      ↓          ↓          ↓
    Cache     Sessions   Anagrams


| Command              | Standard Python?  | What is it?              |
| -------------------- | ----------------  | ------------------------ |
| `dict`               | ✅ Yes            | Built-in data structure  |
| `OrderedDict`        | ✅ Yes            | Standard-library class   |
| `popitem()`          | ✅ Yes            | `OrderedDict` method     |
| `move_to_end()`      | ✅ Yes            | `OrderedDict` method     |
| `get()`              | ✅ Yes            | Dictionary method        |
| `len()`              | ✅ Yes            | Built-in Python function |
| `add_to_lru()`       | ❌ No             | **Our own function**     |
| `get_from_lru()`     | ❌ No             | **Our own function**     |
| `calculate_square()` | ❌ No             | **Our own function**     |

"""
from collections import OrderedDict   #<<-- OrderedDict is a ready-to-use Python class from the standard library.
# Remember which user belongs to a session ID
sessions = {
    "ABC123": "John",
    "XYZ789": "Priya",
    "LMN456": "Rahul"
}
session_id = "XYZ789"
print("\nLogged-in user:", sessions[session_id])

lru_cache = OrderedDict()        # <<---# RECENT USAGE → LRU CACHE
capacity = 3

def add_to_lru(key, value):
    if key in lru_cache:
        lru_cache.move_to_end(key)             # << --     # If already present, move it to the end
    lru_cache[key] = value

    if len(lru_cache) > capacity:              # << --     # If cache is full, remove least recently used item
        removed = lru_cache.popitem(last=False)
        print("Removed:", removed)

def get_from_lru(key):
    if key in lru_cache:
        lru_cache.move_to_end(key)             # << --     # Mark it as recently used
        return lru_cache[key]
    return None

add_to_lru("A", "Google")
add_to_lru("B", "YouTube")
add_to_lru("C", "Amazon")
print("\nLRU Cache:", lru_cache)

# Use A → A becomes recently used
print("Using:", get_from_lru("A"))
print("LRU Cache:", lru_cache)

# Add D → B is removed
add_to_lru("D", "Netflix")
print("Final LRU Cache:", lru_cache)

```
