```python
#----------Graph Representation — Store connections- 
#-- Imagine a social network where people are connected to their friends.
#-- A Hash Map can be used to quickly find "Who/what is connected to this?"

friends = {                              # <<--- Dictionary
    "Rahul": ["Amit", "Priya"],
    "Amit": ["Rahul", "John"],           # <<-- This is list for given dic key
    "Priya": ["Rahul"],
    "John": ["Amit"]
}

for person, connected in friends.items():        # <<-- This is dic loop since there is no [ {}, {} ] type structure, it is {} structure
    print(f"{person} is connected to:")
    for friend in connected:                     # <<-- This loop is for list so no .items() required
        print(f"      ---> ", friend)
```

```python
#----------Database-Style Indexing — Find records quickly

"""
Imagine a hospital has thousands of patient records. Without an index, you might have:
Patient records:

1 → Patient A
2 → Patient B
3 → Patient C
...
100000 → Patient X
---------
A Hash Map can act like an index:
patients = {
    "P1001": {"name": "Rahul", "age": 45},
    "P1002": {"name": "Priya", "age": 38},
    "P1003": {"name": "Amit", "age": 52}
}
"""

students = [
    {"id": 101, "name": "Rahul", "city": "Pune"},          # <<<-- This is list with dic and dic has several key-value pairs 
    {"id": 102, "name": "Priya", "city": "Mumbai"},
    {"id": 103, "name": "Amit", "city": "Delhi"},
    {"id": 104, "name": "Neha", "city": "Pune"},
    {"id": 105, "name": "John", "city": "Bangalore"}
]


student_index = {}                     # << -- # Create an index

# enumerate() is NOT required to create a Hash Map.
# We used it here because we specifically wanted to create an index that maps Student ID → position in the list.
"""
POSITION       STUDENT RECORD
   0      →    {"id": 101, "name": "Rahul", "city": "Pune"}
   1      →    {"id": 102, "name": "Priya", "city": "Mumbai"}
   2      →    {"id": 103, "name": "Amit", "city": "Delhi"}
"""
for position, student in enumerate(students):       #<< -- Here output of enumerate should be position and entire student as a body
    student_index[student["id"]] = position

position = student_index[104]          # << -- # Find student 104
student = students[position]           # << -- # Get the actual record
print(student)
```

```python
# --- Distributed Caching — Data across multiple servers
# Distributed caching means keeping frequently used data in a fast cache that is shared across multiple servers, 
# so the application does not have to repeatedly retrieve the same data from the main database. a large application 
# may have many servers, and the cache can be distributed across multiple machines:
"""
                 Application
                      |
                Distributed Cache
                 /      |      \
                ↓       ↓       ↓
             Cache 1  Cache 2  Cache 3 

# --So, if USER100 or data is stored in Cache 1, the system knows where to find it.
"""

shared_cache = {}
database = {
    "USER100": {
        "name": "Rahul",
        "city": "Pune"
    },
    "USER200": {
        "name": "Priya",
        "city": "Mumbai"
    }
}

def server_request(server_name, user_id):

    if user_id in shared_cache:                                # << --     # STEP 1: Check shared cache
        print("  → Getting data from SHARED CACHE")
        return shared_cache[user_id]
                                # << --  # STEP 2: Cache miss → go to database
    print("  → Going to DATABASE")
    user = database[user_id]

    shared_cache[user_id] = user                               # << -- # STEP 3: Put database result into shared cache
    print("  → Saving data into SHARED CACHE")
    return user

profile = server_request("Server 1", "USER100")                 # << - # Rahul's first request reaches Server 1
print("  → Result:", profile)

profile = server_request("Server 3", "USER100")                 # << -- # Rahul's second request reaches Server 3
print("  → Result:", profile)

profile = server_request("Server 2", "USER100")                 # << -- # Rahul's third request reaches Server 2
print("  → Result:", profile)  
print(shared_cache)
```
