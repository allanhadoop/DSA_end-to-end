```python
# ---You have a shopping list and want to know how many times each item appears.
items = ['paper','spoon', 'brush','cheese','brush','Paper']

count = {}                         # << --- Use dictionary as a temp. storage to update counts
for i in items: 
    inew = i.lower()
    if inew in count:
        count[inew] += 1         # << - Dictionary access is always: using dictionary[key] and not dictionary{key} that is why [inew]
    else:
        count[inew] = 1

print(count)
```

```python
items = ['paper','spoon', 'brush','cheese','brush','Paper']
count = {}
for i in items:
    i = i.lower()
    count[i] = count.get(i,0) + 1          # dictionary.get(key, default_value) here count is dic. 

print(count)
```

```python
# -- You have a list of students and want to quickly check whether a particular student exists.
students = {
    "Rahul": 101,
    "Priya": 102,
    "Amit": 103
}
name = "Priya"
if name in students:        # << --- Look by key
    print("Student exists")
else:
    print("Student not found")
```

```python
# -- Word Frequency - here we're combining text processing + counting.

sentence = "AI is useful and AI is powerful and Python is useful"
words = sentence.split()            # << --- Split function

frequency = {}
for word in words:
    if word in frequency:
        frequency[word] += 1
    else:
        frequency[word] = 1

print(frequency)
```
